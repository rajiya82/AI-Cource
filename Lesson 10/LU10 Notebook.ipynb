# Introduction to APIs

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

An **API** (Application Programming Interface) is a mechanism that allows different software applications to communicate with each other. APIs allow applications to request or provide data and services to other systems, and they are found in virtually every modern software application.

## What is an API Call?

An **API call** is a request sent by a software application to an API, followed by a response from the server. This request–response cycle is the cornerstone of networked API communication, enabling retrieval, creation, modification, and deletion of data.

## Core Concepts

| Concept | Description |
|---|---|
| **Endpoint** | The URL where requests are sent. Example: `https://api.example.com/getWeather` |
| **HTTP Methods** | `GET` (retrieve data), `POST` (send data), `PUT` (update data), `DELETE` (remove data) |
| **Headers** | Additional information sent with the request (e.g., authorisation tokens, content type) |
| **Parameters** | Data included with the request to filter results. Example: `?city=Singapore` |
| **Request Body** | Data sent to the server — used with `POST` and `PUT` requests |
| **Response** | Data returned from the server: a status code, headers, and a body (often JSON) |
| **Authentication** | Verifies permission to access the API — typically via API keys or OAuth tokens |
| **Rate Limiting** | Restrictions on the number of requests per time period to prevent abuse |
| **Error Handling** | Identifying and responding to errors during the API call process |

## HTTP Status Codes

| Code | Meaning |
|---|---|
| `200 OK` | Successful request |
| `201 Created` | Resource created successfully |
| `204 No Content` | Successful, but no content returned |
| `400 Bad Request` | Missing or invalid parameters |
| `401 Unauthorized` | Authentication failed |
| `403 Forbidden` | Authenticated but no access to this resource |
| `404 Not Found` | Resource does not exist |
| `500 Internal Server Error` | Server-side error |
| `503 Service Unavailable` | Server not ready (maintenance or overload) |

#### Sample Call to the API

The code below imports the requests and pandas libraries and defines a function fetch_data that constructs a URL for an API endpoint. It sends an HTTP GET request to the specified URL, passing any optional parameters. If the response status is 200 (successful), it returns the response data as JSON; otherwise, it prints an error message with the status code and returns None.

```python
import requests

def fetch_data(endpoint, params=None):
    """
    Fetch data from the Singapore Government Open Data API.

    Parameters:
    endpoint (str): The API endpoint to fetch data from (e.g., 'air-temperature', 'rainfall').
    params (dict, optional): Additional parameters to pass in the request.

    Returns:
    dict: The JSON response from the API if the request is successful.
    None: If the request fails.
    """
    # Change the base URL to point to the v2 endpoint
    url = f'https://api-open.data.gov.sg/v2/real-time/api/{endpoint}'
    response = requests.get(url, params=params)
    if response.status_code == 200:
        return response.json()
    else:
        print(f'Failed to retrieve data: {response.status_code}')
        return None
```

#### Get the Current Air Temperature and Precipitation Readings

This code uses the `fetch_data` function, previously defined in this notebook, to retrieve temperature and precipitation data from the Singapore Government Open Data API. The `print_data_summary` function then processes this data to identify weather stations that have **both** temperature and precipitation readings, and prints a summary of these stations with their respective temperature and precipitation values. The goal is to provide a concise report of only the stations with complete weather information.

```python
# Function reproduced from above.
def fetch_data(endpoint, params=None):
    """
    Fetch data from the Singapore Government Open Data API.

    Parameters:
    endpoint (str): The API endpoint to fetch data from (e.g., 'air-temperature', 'rainfall').
    params (dict, optional): Additional parameters to pass in the request.

    Returns:
    dict: The JSON response from the API if the request is successful.
    None: If the request fails.
    """
    # Change the base URL to point to the v2 endpoint
    url = f'https://api-open.data.gov.sg/v2/real-time/api/{endpoint}'
    response = requests.get(url, params=params)
    if response.status_code == 200:
        return response.json()
    else:
        print(f'Failed to retrieve data: {response.status_code}')
        return None

# Function to print data summary
def print_data_summary(temperature_data, precipitation_data):
    """
    Print a summary of stations with both temperature and precipitation data.

    Parameters:
    temperature_data (dict): The temperature data retrieved from the API.
    precipitation_data (dict): The precipitation data retrieved from the API.

    The function extracts station information, matches temperature and precipitation readings,
    and prints only the stations where both readings are available.
    """
    # Extract station names from both temperature and precipitation data
    stations = {}
    
    # Process temperature stations
    for station in temperature_data['data']['stations']:
        station_id = station['id']
        stations[station_id] = {'name': station['name'], 'temperature': None, 'precipitation': None}
    
    # Update the dictionary with temperature readings
    temp_items = temperature_data.get('data', {}).get('readings', [])
    for temp_reading in temp_items[0].get('data', []):
        station_id = temp_reading['stationId']
        if station_id in stations:
            stations[station_id]['temperature'] = f"{temp_reading['value']}°C"
    
    # Process precipitation stations (there may be new stations not in temperature data)
    for station in precipitation_data['data']['stations']:
        station_id = station['id']
        if station_id not in stations:
            stations[station_id] = {'name': station['name'], 'temperature': None, 'precipitation': None}
    
    # Update the dictionary with precipitation readings
    precip_items = precipitation_data.get('data', {}).get('readings', [])
    for precip_reading in precip_items[0].get('data', []):
        station_id = precip_reading['stationId']
        if station_id in stations:
            stations[station_id]['precipitation'] = f"{precip_reading['value']}mm"
    
    # Sort stations alphabetically by name
    sorted_stations = sorted(stations.values(), key=lambda x: x['name'])
    
    # Print only stations with both temperature and precipitation readings
    for station in sorted_stations:
        if station['temperature'] is not None and station['precipitation'] is not None:
            print(f"{station['name']}: Temperature: {station['temperature']}, Precipitation: {station['precipitation']}")

# Example usage
temperature_data = fetch_data('air-temperature')
precipitation_data = fetch_data('rainfall')

if temperature_data and precipitation_data:
    print_data_summary(temperature_data, precipitation_data)
```

#### Get Air Temperature Readings for a Specific Date and Time

This code retrieves and prints air temperature readings for a **specific date and time** across Singapore using the Singapore Government Open Data API. It uses the `fetch_data` function to query the API (reproduced from previous cells for convenience), passing a date parameter in the format `YYYY-MM-DD` or `YYYY-MM-DDTHH:MM:SS` to filter results by date or a specific moment in time. The get_air_temperature_for_datetime function then extracts and prints the temperature readings for each station at the given date or date-time, providing a summary of air temperature at multiple locations across Singapore.

```python
# Function reproduced from above.
def fetch_data(endpoint, params=None):
    """
    Fetch data from the Singapore Government Open Data API.

    Parameters:
    endpoint (str): The API endpoint to fetch data from (e.g., 'air-temperature', 'rainfall').
    params (dict, optional): Additional parameters to pass in the request.

    Returns:
    dict: The JSON response from the API if the request is successful.
    None: If the request fails.
    """
    # Change the base URL to point to the v2 endpoint
    url = f'https://api-open.data.gov.sg/v2/real-time/api/{endpoint}'
    response = requests.get(url, params=params)
    if response.status_code == 200:
        return response.json()
    else:
        print(f'Failed to retrieve data: {response.status_code}')
        return None

# Function to get air temperature readings for a specific date and time
def get_air_temperature_for_datetime(date):
    """
    Get air temperature readings for a specific date and time across Singapore.

    Parameters:
    date (str): The date or date-time in SGT for which to retrieve data (YYYY-MM-DD or YYYY-MM-DDTHH:MM:SS).

    Returns:
    None: Prints the station name and temperature readings for the specified date and time.
    """
    params = {'date': date}
    temperature_data = fetch_data('air-temperature', params=params)
    
    if temperature_data:
        # Extract station names and temperature readings
        for station in temperature_data['data']['stations']:
            station_id = station['id']
            station_name = station['name']
            readings = temperature_data.get('data', {}).get('readings', [])
            for temp_reading in readings[0].get('data', []):
                if temp_reading['stationId'] == station_id:
                    temperature = temp_reading['value']
                    print(f"{station_name}: Temperature: {temperature}°C")

# Example usage
date = '2026-04-01T23:59:00'  # Specific date-time
get_air_temperature_for_datetime(date)
```

#### Print More Details from the API.

The code is designed to:

1. Use the `requests` library to send an HTTP GET request to the Singapore government's API for a 4-day weather forecast.  
   
2. If the response status is 200 (OK), it parses the returned JSON data.  
   
3. It then iterates over each item in the 'items' list, and for each item, it iterates over the 'forecasts' list.  
   
4. For each forecast, it extracts certain information such as the date, forecast description, temperature high and low, and relative humidity high and low.  
   
5. This information is collected into a dictionary for each forecast, with the dictionary keys corresponding to the data points mentioned above.  
    
6.  Each dictionary is appended to a list named `results`.  
    
7.  The list of dictionaries (`results`) is then converted into a `pandas` DataFrame for easier display and manipulation.  
    
8.  Finally, the DataFrame is displayed, showing the collected forecast information in a tabular format.  
    
9.  If the API response is not OK, it prints an error message with the failed status code.

```python
import requests
import pandas as pd
from IPython.display import display
import datetime

# Function reproduced from above.
def fetch_data(endpoint, params=None):
    """
    Fetch data from the Singapore Government Open Data API.

    Parameters:
    endpoint (str): The API endpoint to fetch data from (e.g., 'air-temperature', 'rainfall').
    params (dict, optional): Additional parameters to pass in the request.

    Returns:
    dict: The JSON response from the API if the request is successful.
    None: If the request fails.
    """
    # Change the base URL to point to the v2 endpoint
    url = f'https://api-open.data.gov.sg/v2/real-time/api/{endpoint}'    
    response = requests.get(url, params=params)
    if response.status_code == 200:
        return response.json()
    else:
        print(f'Failed to retrieve data: {response.status_code}')
        return None

# Function to get 4-day weather forecast
def get_4_day_weather_forecast(date=None):
    """
    Get the 4-day weather forecast across Singapore.

    Parameters:
    date (str, optional): The date or date-time in SGT for which to retrieve data (YYYY-MM-DD or YYYY-MM-DDTHH:MM:SS).

    Returns:
    None: Displays the 4-day weather forecast information.
    """
    endpoint = 'four-day-outlook'
    params = {'date': date} if date else None
    forecast_data = fetch_data(endpoint, params=params)
    
    if forecast_data:
        # Prepare a list to hold the desired information for each item
        results = []
        today = datetime.datetime.now()
        
        # Iterate through each record in the 'records' list
        for record in forecast_data.get('data', {}).get('records', []):
            for i, forecast in enumerate(record.get('forecasts', [])):
                # Format the forecast information as required
                temperature = forecast.get('temperature', {})
                relative_humidity = forecast.get('relativeHumidity', {})
                wind = forecast.get('wind', {})
                
                # Calculate the correct date for each day of the forecast
                forecast_date = (today + datetime.timedelta(days=i+1)).strftime('%Y-%m-%d')
                
                result = {
                    'Date': forecast_date,
                    'Day': forecast.get('day'),
                    'Temperature': f"Low: {temperature.get('low')}°C, High: {temperature.get('high')}°C" if temperature else 'N/A',
                    'Relative Humidity': f"Low: {relative_humidity.get('low')}%, High: {relative_humidity.get('high')}%" if relative_humidity else 'N/A',
                    'Forecast': forecast.get('forecast', {}).get('summary', 'N/A'),
                    'Wind': f"Low: {wind.get('speed', {}).get('low')} km/h, High: {wind.get('speed', {}).get('high')} km/h, Direction: {wind.get('direction')}" if wind else 'N/A'
                }
                # Add the collected information to the 'results' list
                results.append(result)
        
        # Convert the 'results' list to a DataFrame for better display
        df = pd.DataFrame(results)
        
        # Display the DataFrame using IPython display for better formatting
        display(df)
    else:
        print('Failed to retrieve 4-day weather forecast data')

# Example usage
get_4_day_weather_forecast()
```

```python
import requests

def fetch_weather_forecast(api_key, location, days):
    # Define the API endpoint and parameters
    url = 'https://api.weatherapi.com/v1/forecast.json'
    params = {
        'key': api_key,
        'q': location,
        'days': days
    }

    # Make the API request
    response = requests.get(url, params=params)
    
    # Check for a valid response
    if response.status_code == 200:
        # Parse the JSON response
        return response.json()
    else:
        print(f'Failed to retrieve data: {response.status_code}')
        return None
```

## Matplotlib vs. Seaborn: A Comparison of Python Visualization Libraries

Matplotlib and Seaborn are two widely-used libraries for data visualization in Python, each serving distinct purposes.

Both libraries complement each other, allowing users to leverage Seaborn’s simplicity while accessing Matplotlib's customization capabilities when needed.

# Visualising API Data

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

Combining API calls with visualisations enables real-time data retrieval and dynamic representation, which enhances understanding and facilitates decision-making by revealing patterns and insights that might be missed in raw data.

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Function to fetch data from the API
def fetch_data(endpoint, params=None):
    url = f'https://api.data.gov.sg/v1/environment/{endpoint}'
    response = requests.get(url, params=params)
    if response.status_code == 200:
        return response.json()
    else:
        print(f'Failed to retrieve data: {response.status_code}')
        return None

# Function to plot temperature and relative humidity
def plot_temperature_humidity(data):
    if data:
        results = []
        dates = []
        temp_high = []
        temp_low = []
        humidity_high = []
        humidity_low = []
        
        for day in data.get('items', [])[0].get('forecasts', []):
            result = {
                'Date': day.get('date'),
                'Forecast': day.get('forecast'),
                'Temperature Low (°C)': day.get('temperature', {}).get('low'),
                'Temperature High (°C)': day.get('temperature', {}).get('high'),
                'Relative Humidity Low (%)': day.get('relative_humidity', {}).get('low'),
                'Relative Humidity High (%)': day.get('relative_humidity', {}).get('high'),
            }
            results.append(result)
            dates.append(day.get('date'))
            temp_high.append(day.get('temperature', {}).get('high'))
            temp_low.append(day.get('temperature', {}).get('low'))
            humidity_high.append(day.get('relative_humidity', {}).get('high'))
            humidity_low.append(day.get('relative_humidity', {}).get('low'))
        
        df = pd.DataFrame(results)
        display(df)
        
        fig, ax1 = plt.subplots(figsize=(10, 5))
        
        # Plot temperature data
        ax1.plot(dates, temp_high, label='High Temperature (°C)', marker='o', color='tab:red')
        ax1.plot(dates, temp_low, label='Low Temperature (°C)', marker='o', color='tab:orange')
        ax1.set_xlabel('Date')
        ax1.set_ylabel('Temperature (°C)', color='tab:red')
        ax1.tick_params(axis='y', labelcolor='tab:red')
        ax1.legend(loc='upper left')
        ax1.grid(True)
        
        # Instantiate a second axes that shares the same x-axis
        ax2 = ax1.twinx()
        
        # Plot humidity data
        ax2.plot(dates, humidity_high, label='High Relative Humidity (%)', marker='o', color='tab:blue')
        ax2.plot(dates, humidity_low, label='Low Relative Humidity (%)', marker='o', color='tab:cyan')
        ax2.set_ylabel('Relative Humidity (%)', color='tab:blue')
        ax2.tick_params(axis='y', labelcolor='tab:blue')
        ax2.legend(loc='upper right')
        
        plt.title('4-Day Weather Forecast for Singapore')
        
        # Place the legends below the graph
        ax1.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), fancybox=True, shadow=True, ncol=2)
        ax2.legend(loc='upper center', bbox_to_anchor=(0.5, -0.25), fancybox=True, shadow=True, ncol=2)
        
        fig.tight_layout()  # to ensure the right y-label is not clipped
        plt.show()
    else:
        print("No data available")

# Fetch the 4-day weather forecast data
data = fetch_data('4-day-weather-forecast')

# Call the function to plot the temperature and humidity forecast
plot_temperature_humidity(data)
```

Now we will use our skills in pandas and string manipulation to illustrate the price in a more readable format.

```python
import requests
import pandas as pd
from datetime import datetime, timezone

def get_bitcoin_price():
    url = "https://api.coingecko.com/api/v3/simple/price"
    params = {
        'ids': 'bitcoin',
        'vs_currencies': 'USD',
        'include_market_cap': 'true',
        'include_24hr_vol': 'true',
        'include_24hr_change': 'true',
        'include_last_updated_at': 'true',
        'precision': 'full'
    }
    response = requests.get(url, params=params)
    data = response.json()
    return data

def format_data(data):
    bitcoin_info = {
        'Price (USD)': f"${data['bitcoin']['usd']:.2f}",
        'Market Cap (USD)': f"${data['bitcoin']['usd_market_cap']:.2f}",
        '24hr Volume (USD)': f"${data['bitcoin']['usd_24h_vol']:.2f}",
        '24hr Change (%)': f"{data['bitcoin']['usd_24h_change']:.2f}%",
        'Last Updated': datetime.fromtimestamp(data['bitcoin']['last_updated_at'], timezone.utc).strftime('%d-%b-%Y %H:%M')
    }
    df = pd.DataFrame([bitcoin_info])
    return df

# Usage:
bitcoin_data = get_bitcoin_price()
formatted_data = format_data(bitcoin_data)
display(formatted_data)
```

Graphing the response using Matplotlib and SeaBorn

```python
import requests
import json
from datetime import datetime
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# API request
url = "https://api.coingecko.com/api/v3/coins/bitcoin/market_chart?vs_currency=usd&days=30&interval=daily&precision=2"
headers = {"accept": "application/json"}
response = requests.get(url, headers=headers)

# Convert the response to JSON format
data = response.json()

# Extract prices (timestamps and values)
prices = data.get("prices", [])

# Create lists for dates and closing prices
dates = []
closing_prices = []

for price in prices:
    timestamp = price[0]
    closing_price = price[1]
    
    # Convert timestamp (in milliseconds) to readable date
    date = datetime.fromtimestamp(timestamp / 1000).strftime('%Y-%m-%d')
    
    # Append to lists
    dates.append(date)
    closing_prices.append(closing_price)

# Create a DataFrame using Pandas
df = pd.DataFrame({
    'Date': dates,
    'Closing Price (USD)': closing_prices
})

# Set the 'Date' column as the DataFrame index for better plotting
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)

# Plot using Seaborn
plt.figure(figsize=(10,6))
sns.lineplot(data=df, x=df.index, y='Closing Price (USD)')
plt.title('Bitcoin Closing Prices Over the Last 30 Days')
plt.xlabel('Date')
plt.ylabel('Closing Price (USD)')
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.tight_layout()

# Show the plot
plt.show()
```
