# What is OpenCode?

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

OpenCode is a **terminal-based AI coding agent** that runs directly inside your VS Code integrated terminal. It reads your project files, understands your codebase, and can plan and implement changes on your behalf.

| Feature | Description |
|---|---|
| **Terminal-native** | Runs in the VS Code integrated terminal — no separate browser tab or app required |
| **File-aware** | Reads the files in your project and uses them as context for every prompt |
| **Two modes** | Plan mode for thinking through changes; Build mode for making them |
| **Provider-agnostic** | Works with GitHub Copilot, Anthropic, OpenAI, and other model providers |
| **File references** | Use `@filename` in any prompt to pin a specific file into context |
| **Undo support** | `/undo` reverts the last change and restores the original prompt |

## Where OpenCode Fits in a VS Code Workflow

You write and review code in the editor. OpenCode runs in the integrated terminal panel below it. When you give it a task, it reads your files, applies changes, and you review the result in the editor — all without switching windows.

```
┌─────────────────────────────────────┐
│  VS Code Editor                     │  ← OpenCode reads and edits files here
├─────────────────────────────────────┤
│  VS Code Integrated Terminal        │  ← You type prompts here
└─────────────────────────────────────┘
```

# Node.js Installation

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

OpenCode is distributed as an npm package and requires Node.js. Before installing OpenCode, check whether Node.js is already present on your machine.

## Step 1 — Check if Node.js is Installed

Open the **VS Code integrated terminal** (`` Ctrl+` `` on Windows and Mac) and run both commands:

```bash
node --version
npm --version
```

If both return a version number (for example `v22.14.0` and `10.9.2`), Node.js is already installed. Skip to [Installing OpenCode](#Installing-OpenCode).

If you see `command not found` or a similar error, follow the steps below for your operating system.

## Step 2 — Install Node.js on Windows

The recommended method on Windows is **winget**, the built-in package manager. Run this in the terminal:

```bash
winget install OpenJS.NodeJS.LTS
```

After the installer finishes, **close and reopen the terminal**, then verify:

```bash
node --version
npm --version
```

If winget is not available, download the LTS installer directly from [nodejs.org](https://nodejs.org).

## Step 2 — Install Node.js on macOS

The recommended method on macOS is **Homebrew**. If Homebrew is not yet installed, run:

```bash
/bin/bash -c "$(curl -fsSL https://raw.githubusercontent.com/Homebrew/install/HEAD/install.sh)"
```

Once Homebrew is available:

```bash
brew install node
```

Verify after installation:

```bash
node --version
npm --version
```

## Referencing Files with @

Prefix a filename with `@` inside any prompt to add that file directly to the context:

```
@weather.py explain what this script does
```

OpenCode uses fuzzy matching, so `@weather` is sufficient if the filename is unambiguous. You can reference multiple files in a single prompt:

```
@weather.py @config.py update the API timeout setting
```

# Weather App — Starter Project

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

The project for this lesson is a small Python script that fetches and displays the current weather for a city the user chooses. It uses the **wttr.in** service, which requires no API key or account.

The script:
- Asks the user to enter a city name
- Calls `https://wttr.in/{city}?format=j1` and parses the JSON response
- Displays the location, current date and time, weather condition, temperature, humidity, and wind speed

## Installing the requests Library

In a terminal install requests. First ensure you have activated the environment `ai-course`. Recommended to use CMD not POWERSHELL.

````python
conda activate ai-course
conda install requests
````

## Creating weather.py

The `%%writefile` magic command saves the cell content directly to `weather.py` in the current directory. Running the cell writes the file but does not execute the script.

````python
%%writefile weather.py
import requests
from datetime import datetime


def get_weather(city):
    url = f"https://wttr.in/{city}?format=j1"
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        data = response.json()

        current = data["current_condition"][0]
        area = data["nearest_area"][0]

        location = area["areaName"][0]["value"]
        country = area["country"][0]["value"]
        temp_c = current["temp_C"]
        feels_like = current["FeelsLikeC"]
        description = current["weatherDesc"][0]["value"]
        humidity = current["humidity"]
        wind_kmph = current["windspeedKmph"]

        now = datetime.now()

        print(f"\nWeather Report")
        print(f"{'=' * 40}")
        print(f"Location  : {location}, {country}")
        print(f"Date/Time : {now.strftime('%d %b %Y, %H:%M')}")
        print(f"Condition : {description}")
        print(f"Temp      : {temp_c}\u00b0C (feels like {feels_like}\u00b0C)")
        print(f"Humidity  : {humidity}%")
        print(f"Wind      : {wind_kmph} km/h")

    except requests.exceptions.RequestException as e:
        print(f"Error fetching weather data: {e}")
    except (KeyError, IndexError) as e:
        print(f"Error parsing weather data: {e}")


if __name__ == "__main__":
    city = input("Enter a city name: ")
    get_weather(city)
````

## Running the Starter Script

To run the script, open the **VS Code integrated terminal** and type:

```bash
python weather.py
```

Try a city such as **Singapore**, **London**, or **Tokyo**. Confirm that a weather report appears before moving on.

Example output:

## Using @ for File References

The `@` prefix pins a specific file into the conversation context. Without it, OpenCode infers which files are relevant from the project, but using `@` makes it explicit.

| Prompt style | Behaviour |
|---|---|
| `@weather.py explain ...` | OpenCode reads `weather.py` before responding |
| `explain the weather app` | OpenCode infers context from the project |
| `@weat` | Fuzzy match — resolves to `weather.py` if unambiguous |

# Plan Mode

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

Plan mode is used to think through a change **before any code is written**. In Plan mode, file write, edit, and bash tool access are disabled — OpenCode can only read files and respond with plans.

## Entering Plan Mode

Press `Tab` inside the OpenCode prompt to toggle between Build and Plan mode. The mode indicator in the interface updates to confirm the switch.

## Planning a New Feature

Switch to Plan mode and enter this prompt:

```
@weather.py I want to add a 3-day forecast below the current conditions.
Each day should show the date, a short weather description, and the high
and low temperatures in Celsius. Show me a plan before making any changes.
```

## Reviewing the Plan

OpenCode responds with a step-by-step breakdown of what it intends to do. Read through it and check:

- Does it identify the correct part of the code to change?
- Does it describe the new output format clearly?
- Is anything missing or unclear?

## Refining with Follow-up Prompts

If the plan is incomplete, ask a follow-up question while still in Plan mode. For example:

```
The forecast data is in data["weather"]. Each entry has "date", "maxtempC",
"mintempC", and the midday condition is in hourly[4]["weatherDesc"][0]["value"].
Update your plan with this structure.
```

Stay in Plan mode until you are satisfied with the approach. No files will be changed until you switch to Build mode.

# Exit Task Answers

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

## Example Prompts

### Plan mode prompt

```
@weather.py I want to add a unit toggle. Before fetching the weather,
ask the user to enter C for Celsius or F for Fahrenheit and store the
choice. Add a helper function that converts a Celsius value to
Fahrenheit using the formula F = (C * 9/5) + 32. Apply the conversion
to temp_C, FeelsLikeC, maxtempC, and mintempC before printing.
Default to Celsius if the user enters anything other than F.
Show me the plan before making any changes.
```

### Build mode prompt

```
implement the plan
```

````python
import requests
from datetime import datetime


def convert_temp(temp_c, unit):
    if unit == "F":
        return round((float(temp_c) * 9 / 5) + 32, 1)
    return float(temp_c)


def get_weather(city, unit="C"):
    url = f"https://wttr.in/{city}?format=j1"
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        data = response.json()

        current = data["current_condition"][0]
        area = data["nearest_area"][0]

        location = area["areaName"][0]["value"]
        country = area["country"][0]["value"]
        temp = convert_temp(current["temp_C"], unit)
        feels_like = convert_temp(current["FeelsLikeC"], unit)
        description = current["weatherDesc"][0]["value"]
        humidity = current["humidity"]
        wind_kmph = current["windspeedKmph"]

        now = datetime.now()

        print(f"\nWeather Report")
        print(f"{'=' * 40}")
        print(f"Location  : {location}, {country}")
        print(f"Date/Time : {now.strftime('%d %b %Y, %H:%M')}")
        print(f"Condition : {description}")
        print(f"Temp      : {temp}\u00b0{unit} (feels like {feels_like}\u00b0{unit})")
        print(f"Humidity  : {humidity}%")
        print(f"Wind      : {wind_kmph} km/h")

        print(f"\n3-Day Forecast")
        print(f"{'=' * 40}")
        for day in data["weather"]:
            date = day["date"]
            high = convert_temp(day["maxtempC"], unit)
            low = convert_temp(day["mintempC"], unit)
            condition = day["hourly"][4]["weatherDesc"][0]["value"]
            print(f"{date} | {condition:20s} | High: {high}\u00b0{unit}  Low: {low}\u00b0{unit}")

    except requests.exceptions.RequestException as e:
        print(f"Error fetching weather data: {e}")
    except (KeyError, IndexError) as e:
        print(f"Error parsing weather data: {e}")


if __name__ == "__main__":
    city = input("Enter a city name: ")
    unit = input("Display temperatures in Celsius or Fahrenheit? (C/F): ").strip().upper()
    if unit not in ("C", "F"):
        unit = "C"
    get_weather(city, unit)
````
