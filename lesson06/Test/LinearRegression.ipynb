```python
w = 200
b = 100

def compute_model_output(x, w, b):
    m = x.shape[0]
    f_wb = np.zeros(m)
    print(f_wb)
    for i in range(m):
        f_wb[i] = x[i] * w + b
    print(f_wb)    
    return f_wb


temp_f_wb = compute_model_output(x_train, w, b)
plt.plot(x_train, temp_f_wb, c='b')
plt.scatter(x_train, y_train, c='r')

plt.xlabel = 'size in sqm'
plt.ylabel = 'price in 1000 nds'
plt.title = 'House Price'
# plt.legend()
plt.show()
    
```
