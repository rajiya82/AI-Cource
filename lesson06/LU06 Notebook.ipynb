Functions in Python are blocks of code designed to do one specific job. They allow for **code modularity**, making your code more organised, reusable, and easy to understand. Functions are a fundamental concept in programming, promoting **code reuse** and **procedural decomposition**.

Here is a breakdown of the key components:

- **`def`**: The keyword that signals the start of a function definition.
- **`function_name`**: A descriptive name, following the `snake_case` naming convention.
- **`parameters`**: Values the function accepts as input, placed in parentheses after the name.
- **`:`**: A colon at the end of the function header marks the start of the function body.
- **`statement(s)`**: The indented instructions the function will execute.
- **`return`**: Exits the function and returns a value to the caller.

```python
def function_name(parameters):
    # statement(s)
    return value
```

The following examples illustrate function definition and calling, starting from the least complex and gradually increasing in complexity.

````python
# Example 1: Simple function with no parameters
def say_hello():
    print("Hello!")

say_hello()

# Output: Hello!
````

````python
# Example 2: Function with a parameter
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")

# Output: Hello, Alice!
````

````python
# Example 3: Function with a return value
def add(a, b):
    return a + b

sum_result = add(5, 3)
print(sum_result)

# Output: 8
````

## Functions — Scenario to Code

**Scenario:** A coffee shop wants to automate the process of taking orders and computing the total price. They sell two types of coffee:

- Regular ($5)
- Latte ($7)

They offer a discount of 10% for orders of 10 or more coffees.

Write a Python program with **two functions**: one to calculate the total price of the coffee order, and another to apply the discount if applicable.

No flowchart is provided. If a flowchart will help your understanding, please construct one with [Draw.io](https://app.diagrams.net).

````python
def calculate_total_price(regular_coffees, latte_coffees):
    """
    Calculates the total price of the coffee order.

    Parameters:
        regular_coffees (int): The number of regular coffees ordered.
        latte_coffees (int): The number of latte coffees ordered.

    Returns:
        float: The total price of the coffee order.
    """
    regular_price = 5
    latte_price = 7
    total_price = (regular_coffees * regular_price) + (latte_coffees * latte_price)
    return total_price

def apply_discount(total_price):
    """
    Applies a 10% discount to the total price.

    Parameters:
        total_price (float): The total price of the coffee order.

    Returns:
        float: The total price after the discount.
    """
    discount_rate = 0.1
    discounted_price = total_price * (1 - discount_rate)
    return discounted_price

# Input: number of regular and latte coffees ordered
regular_coffees = int(input("Enter the number of regular coffees: "))
latte_coffees = int(input("Enter the number of latte coffees: "))

# Determine the total price
total_price = calculate_total_price(regular_coffees, latte_coffees)

# Apply discount if applicable
if regular_coffees + latte_coffees >= 10:
    total_price = apply_discount(total_price)

# Output the total price
print(f"The total price is ${total_price:.2f}")
````

Lambda functions, often referred to as **lambdas**, are **small anonymous functions** in Python. They can have any number of parameters but can only have **one expression**. Lambdas are concise and are often used for short, simple operations where writing a standard function would be more verbose.

Here is a breakdown of the key components:

- **`lambda`**: The keyword that signals the start of a lambda function.
- **`parameters`**: Values the lambda accepts as input. Parentheses are optional.
- **`:`**: A colon separates the parameters from the expression.
- **`expression`**: The single expression the lambda evaluates and returns.

```python
lambda parameters: expression
```

The following examples illustrate lambda functions, starting from the least complex and gradually increasing in complexity.

````python
# Example 1: Lambda to double a number
double = lambda x: x * 2
result = double(4)
print(result)

# Output: 8
````

````python
# Example 2: Lambda with multiple parameters
add = lambda a, b: a + b
sum_result = add(5, 3)
print(sum_result)

# Output: 8
````

````python
# Example 3: Lambda as an argument to map()
arr = [1, 2, 3, 4]
doubled_arr = list(map(lambda x: x * 2, arr))
print(doubled_arr)

# Output: [2, 4, 6, 8]
````

## Example 4: Lambda with `sorted()` and `key=`

The most common real-world use of lambdas is as the `key=` argument to `sorted()`. The `key=` argument tells Python *how* to sort — you pass it a function that takes one item and returns the value to sort by.

A lambda is ideal here because the sort criterion is usually a short, one-off expression — not something worth defining as a full function:

```python
sorted(iterable, key=lambda item: item["field"])
```

This pattern comes up constantly when working with lists of dictionaries — for example, sorting products by price, sorting students by score, or sorting employees by name.

````python
products = [
    {"name": "Laptop",     "price": 1299.00},
    {"name": "Mouse",      "price":   29.99},
    {"name": "Keyboard",   "price":   79.00},
    {"name": "Monitor",    "price":  399.00},
]

# Sort by price (cheapest first)
by_price = sorted(products, key=lambda p: p["price"])
for p in by_price:
    print(f"{p["name"]:12s}  ${p["price"]:>8.2f}")

print()

# Sort by name (alphabetical)
by_name = sorted(products, key=lambda p: p["name"])
for p in by_name:
    print(f"{p["name"]:12s}  ${p["price"]:>8.2f}")

print()

# Sort by price descending (most expensive first)
by_price_desc = sorted(products, key=lambda p: p["price"], reverse=True)
print("Most expensive:", by_price_desc[0]["name"])
print("Least expensive:", by_price_desc[-1]["name"])

# Output:
# Mouse           $   29.99
# Keyboard        $   79.00
# Monitor         $  399.00
# Laptop          $ 1299.00
#
# Keyboard        $   79.00
# Laptop          $ 1299.00
# Monitor         $  399.00
# Mouse           $   29.99
#
# Most expensive: Laptop
# Least expensive: Mouse
````

**Scenario:** A coffee shop sells Regular ($5) and Latte ($7) coffees. A 10% discount applies to orders of 10 or more coffees.

In this version, the `calculate_total_price` function has been replaced with a lambda. The `apply_discount` function remains unchanged.

````python
# Lambda to calculate the total price of the coffee order
calculate_total_price = lambda regular_coffees, latte_coffees: (regular_coffees * 5) + (latte_coffees * 7)

def apply_discount(total_price):
    """
    Applies a 10% discount to the total price.

    Parameters:
        total_price (float): The total price of the coffee order.

    Returns:
        float: The total price after the discount.
    """
    discount_rate = 0.1
    discounted_price = total_price * (1 - discount_rate)
    return discounted_price

# Input: number of regular and latte coffees ordered
regular_coffees = int(input("Enter the number of regular coffees: "))
latte_coffees = int(input("Enter the number of latte coffees: "))

# Determine the total price using the lambda
total_price = calculate_total_price(regular_coffees, latte_coffees)

# Apply discount if applicable
if regular_coffees + latte_coffees >= 10:
    total_price = apply_discount(total_price)

# Output the total price
print(f"The total price is ${total_price:.2f}")
````

In the following two cells, we have two implementations that produce the same output. The first uses a standard function, the second uses a lambda. You will notice they are similar, but the lambda is more concise.

| | Function | Lambda |
|---|---|---|
| **Keyword** | `def` | `lambda` |
| **Name** | Required | Anonymous (optional assignment) |
| **Body** | Multiple statements allowed | Single expression only |
| **Return** | Explicit `return` | Implicit (expression result) |
| **Best for** | Complex, reusable logic | Short, one-off operations |

````python
# Example of a void function to print 12x table
def fn_12x_table():
    for i in range(1, 13):
        print(f"12 x {i} = {12 * i}")

# Calling the void function
print("Using void function to print 12x table:")
fn_12x_table()
````

````python
# Example of a lambda function to calculate 12x table
lambda_12x_table = lambda x: 12 * x

# Calling the lambda function repetitively to print 12x table
print("\nUsing lambda function to print 12x table:")
for i in range(1, 13):
    print(f"12 x {i} = {lambda_12x_table(i)}")
````

# Common Mistakes

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

Navigating functions and lambdas for the first time can lead to subtle errors that are challenging to pinpoint. Below is a summary of the most common mistakes.

| Category | Mistake | Notes |
|---|---|---|
| **Syntax** | Missing colon after function header | `def f()` → `def f():` |
| **Name** | Misspelling the function name on call | `fuction2()` → `function2()` |
| **Parameters** | Wrong number of arguments | Calling `function3()` when it expects one argument |
| **Return** | Forgetting `return` | Expression in body with no `return` yields `None` |
| **Scope** | Modifying a global variable without `global` | `x += 5` inside function raises `UnboundLocalError` |
| **Lambda** | Missing comma between parameters | `lambda x y: x + y` → `lambda x, y: x + y` |

The code cell below demonstrates each mistake. Commented-out lines show the incorrect version; uncommented lines show the correct version. Use this as a reference when debugging your own code.

````python
# Syntax Error: Missing colon
# def function1()          # Incorrect
def function1():            # Correct
    return "Hello, World!"

# Name Error: Misspelling the function name
def function2():
    return "Hello, World!"
# fuction2()               # Incorrect — typo
function2()                 # Correct

# Parameter Error: Missing argument
def function3(arg):
    return arg
# function3()              # Incorrect — no argument provided
function3("hello")          # Correct

# Return Mistake: Forgetting return
def function4():
    # "Hello, World!"      # Incorrect — no return, returns None
    return "Hello, World!" # Correct

# Scope Error: Modifying global variable without global keyword
x = 10
def function5():
    global x
    x += 5                 # Correct with global declaration
    return x

# Lambda Error: Missing comma between parameters
# lamba_function = lambda x y: x + y   # Incorrect — missing comma
lambda_function = lambda x, y: x + y   # Correct

print(function1())
print(function3("test"))
print(lambda_function(5, 3))
````
