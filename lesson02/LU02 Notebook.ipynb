A variable is an **identifier (name)** that refers to a **value stored in memory**. It represents a **storage location** in a computer program that holds **data**. In layman's terms, a variable is a **named container** that **stores data** so a program can **use and change it later**.

Variables allow developers to **store**, **retrieve**, and **manipulate data** in their programs.

## Data Types

A **data type** defines what kind of **data** a variable can store and how that data can be **used or processed**. In layman's terms, it tells the computer whether something is a **number, text, or another kind of value**, so it knows what it can do with it.

1. **Integers**: Represents whole numbers without decimal points.
2. **Floating** point numbers: Represents real numbers, i.e., numbers with decimal points.
3. **Strings**: Represents sequences of characters enclosed in quotes.

# Variable Manipulation

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

Python provides various operations to manipulate variables and perform calculations.

## Variable Operators

Addition (`+`), Subtraction (`-`), Multiplication (`*`), Division (`/`), Exponentiation (`**`), Floor Division (`//`), Modulus (`%`)

## Operator Precedence

In order:

1. `()`
2. `**`
3. `*, /, //, %`
4. `+, -`

PEMDAS is a common acronym to remember the order of operations: Parentheses, Exponents, Multiplication and Division (from left to right), Addition and Subtraction (from left to right).

### Operator Descriptions

| Operator | Name | Description | Example |
|---|---|---|---|
| `+` | Addition | Adds two values | `3 + 2 = 5` |
| `-` | Subtraction | Subtracts the right from the left | `3 - 2 = 1` |
| `*` | Multiplication | Multiplies two values | `3 * 2 = 6` |
| `/` | Division | Divides and returns a float | `7 / 2 = 3.5` |
| `//` | Floor Division | Divides and rounds down to an integer | `7 // 2 = 3` |
| `**` | Exponentiation | Raises to a power | `2 ** 3 = 8` |
| `%` | Modulus | Returns the remainder after division | `7 % 2 = 1` |

````python
# Basic arithmetic operations
a = 10
b = 3

print(f"Addition: {a} + {b} = {a + b}")          # Addition
print(f"Subtraction: {a} - {b} = {a - b}")       # Subtraction
print(f"Multiplication: {a} * {b} = {a * b}")    # Multiplication
print(f"Division: {a} / {b} = {a / b}")          # Division (returns float)
print(f"Floor Division: {a} // {b} = {a // b}")  # Floor division (returns int)
print(f"Modulus: {a} % {b} = {a % b}")           # Modulus (remainder)
print(f"Exponentiation: {a} ** {b} = {a ** b}") # Exponentiation

# Order of operations (PEMDAS: Parentheses, Exponents, Multiplication/Division, Addition/Subtraction)
result = (2 + 3) * 4  
print(f"(2 + 3) * 4 = {result}")

# Output: 
# Addition: 10 + 3 = 13
# Subtraction: 10 - 3 = 7
# Multiplication: 10 * 3 = 30
# Division: 10 / 3 = 3.3333333333333335
# Floor Division: 10 // 3 = 3
# Modulus: 10 % 3 = 1
# Exponentiation: 10 ** 3 = 1000
# (2 + 3) * 4 = 20
````

Python's `datetime` module provides classes for working with dates and times.

### Creating Time Variables

Use the `datetime` module.

### Adding Time

Use `timedelta` to add to a given time.

### Estimating Time Difference

Here's a slightly more advanced example. Look through carefully to fully understand the functionality.

> **Note**: The `print()` function can only concatenate strings with other strings. Use `str()` to convert numeric values when building output strings, or use f-strings.

### Getting Today's Date

Use `date.today()` from the `datetime` module to get the current date as a `date` object (year, month, day only, without time). This is useful when you only need the calendar date rather than a full timestamp.

````python
from datetime import datetime 

# Get current date and time
current_time = datetime.now() 
print("Current date and time:", current_time)
print(f"Year: {current_time.year}, Month: {current_time.month}, Day: {current_time.day}")
print(f"Hour: {current_time.hour}, Minute: {current_time.minute}, Second: {current_time.second}")

# Format the date and time
formatted_date = current_time.strftime("%Y-%m-%d %H:%M:%S")
print("Formatted date and time:", formatted_date)

# Output (as of April 15, 2026):
# 2026-04-15 10:30:00.000000
````

````python
from datetime import timedelta

# Adding time to a datetime
tomorrow = current_time + timedelta(days=1)
print(f"Tomorrow: {tomorrow}")

one_week_later = current_time + timedelta(weeks=1)
print(f"One week later: {one_week_later}")

# Subtracting time
yesterday = current_time - timedelta(days=1)
print(f"Yesterday: {yesterday}")

# Output:
# Tomorrow: 2026-04-16 10:30:00.000000
````

````python
from datetime import date

# Create a specific date
date_2001 = date(2001, 1, 1)  # January 1, 2001 (Space Odyssey)
print(f"Space Odyssey date: {date_2001}")

# Get the current date
current_date = date.today()
print(f"Today's date: {current_date}")

# Calculate the difference in days
difference_in_days = (current_date - date_2001).days
print(f"{difference_in_days} days since 01/01/2001 (Space Odyssey)")

# Calculate years, months and remaining days
years = difference_in_days // 365
remaining_days = difference_in_days % 365
print(f"That's approximately {years} years and {remaining_days} days")

# Output (as of April 15, 2026):
# Space Odyssey date: 2001-01-01
# Today's date: 2026-04-15
# 9235 days since 01/01/2001 (Space Odyssey)
# That's approximately 25 years and 110 days
````

Understanding common error types helps with debugging and writing more robust code.

### TypeError

A `TypeError` occurs when an operation or function is applied to an object of an inappropriate type. It is a way for Python to tell you that you're trying to do something that does not make sense given the type of data you have.

### Checking Data Type

It's important to know the type of data you're working with, especially when troubleshooting errors. You can check the data type of an object using the `type()` function.

### Conversion Functions

Sometimes, you'll want to convert data from one type to another. Python provides several functions:

- `int()` converts a value to an integer.
- `float()` converts a value to a floating-point number.
- `str()` converts a value to a string.

> **Note**: Converting between types can sometimes cause data loss, as seen in the conversion of float to int, so it's important to be aware of the potential pitfalls.

````python
# Type Error Example
try:
    # This will cause a TypeError because you can't add a string and an integer together
    result = "hello" + 5
    print(result)
except TypeError as e:
    print(f"TypeError caught: {e}")

# Output: TypeError caught: can only concatenate str (not "int") to str
````

# Comparison Operators

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

Comparison operators are used to compare values. They return `True` or `False` according to the condition. Here are the common comparison operators in Python:

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 4` | `True` |
| `<` | Less than | `4 < 5` | `True` |
| `>` | Greater than | `5 > 4` | `True` |
| `<=` | Less than or equal to | `5 <= 5` | `True` |
| `>=` | Greater than or equal to | `5 >= 4` | `True` |

### Assignment vs Comparison

It's crucial to differentiate between assignment and comparison, as they serve different purposes.

- `=` **Assignment**: stores the value on the right into the variable on the left. Example: `age = 18` assigns the value `18` to the variable `age`.
- `==` **Comparison**: checks if the values on either side are equal. Example: `age == 18` returns `True` if `age` is `18`, `False` otherwise.

### Decision Making with Comparison Operators

In decision-making structures, comparison operators are often used to evaluate conditions. The `if` statement is used to specify a block of code to be executed if a condition is true.

````python
# Comparison examples
print(5 == 5)   # True
print(5 != 4)   # True
print(4 < 5)    # True
print(5 > 4)    # True
print(5 <= 5)   # True
print(5 >= 4)   # True

# Assignment vs comparison
age = 18          # assigns 18 to age
print(age == 18)  # True - compares age to 18
print(age == 19)  # False
````

# Decision Making

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

Conditional statements allow your program to make decisions based on conditions.

In the code snippet below:

1. The `if` statement checks the condition `age >= 18`.
2. If the condition is `True`, it executes the code within the `if` block, printing "Adult".
3. If the condition is `False`, it executes the code within the `else` block, printing "Minor".

This way, Python allows for decision-making processes based on the evaluation of conditions, enabling more complex and dynamic code execution.

The `elif` keyword is shorthand for `else if`. It allows multiple conditions to be checked in sequence. Here's an example utilising `if`, `elif`, and `else` to categorise different age groups:

In the `elif` example:

1. An `age` variable is defined with a value of `25`.
2. The `if` statement checks whether `age` is less than `13`.
3. The `elif` statements check additional ranges in sequence.
4. The `else` statement handles any remaining case.
5. Finally, the `print` statement outputs the category based on the age value.

````python
# Basic if-else statement
age = 19

if age >= 18:
    print("Adult")
else:
    print("Minor")

# Output: Adult
````

````python
# More complex if-elif-else chain
age = 25

if age < 13:
    category = "Child"
elif 13 <= age < 18:
    category = "Teenager"
elif 18 <= age < 65:
    category = "Adult"
else:
    category = "Senior"

print(f"You are categorized as: {category}")

# Output: You are categorized as: Adult
````

````python
# Conditional expressions (ternary operator)
age = 20
status = "Adult" if age >= 18 else "Minor"
print(f"Status: {status}")

# Nested conditions
num = 15
if num > 0:
    if num % 2 == 0:
        print(f"{num} is a positive even number")
    else:
        print(f"{num} is a positive odd number")
elif num < 0:
    print(f"{num} is negative")
else:
    print(f"{num} is zero")
    
# Output: 
# Status: Adult
# 15 is a positive odd number
````

# IPO Revisit

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

The IPO model stands for **Input-Process-Output**, and it is a fundamental concept in programming and problem-solving.

### Input

The first step in the process is gathering the required data from the user. The `input()` function is used to prompt the user to enter their data. The `float()` function converts the user's input to a floating-point number, as `input()` returns a string by default.

### Process

The next step is to process the input data to obtain the desired output. In the BMI example, the Body Mass Index is calculated using the formula `BMI = weight / (height ** 2)`. The expression `height ** 2` calculates the square of the height.

### Output

Finally, the result of the process is displayed to the user using the `print()` function.

By following the IPO model, the code is organised in a logical and clear manner, making it easier to understand and debug. Each step of the process is clearly delineated, and the flow of data from input to process to output is straightforward.

Let's implement a BMI (Body Mass Index) calculator using the IPO model:

````python
# INPUT
height = float(input("Enter your height in meters: "))
weight = float(input("Enter your weight in kg: "))
````

````python
# PROCESS
bmi = weight / (height ** 2)

# Determine BMI category
if bmi < 18.5:
    category = "Underweight"
elif 18.5 <= bmi < 25:
    category = "Normal weight"
elif 25 <= bmi < 30:
    category = "Overweight"
else:
    category = "Obesity"
````

````python
# OUTPUT
print(f"Your BMI is: {bmi:.2f}")
print(f"This is categorized as: {category}")

# Output (for height=1.8m and weight=80kg):
# Your BMI is: 24.69
# This is categorized as: Normal weight
````

### IPO Model: Visual Representation

```
+-------------+     +------------------+     +----------------+
|   INPUT     |     |     PROCESS      |     |     OUTPUT     |
| ----------- | --> | ---------------- | --> | -------------- |
| height (m)  |     | Calculate BMI    |     | BMI value      |
| weight (kg) |     | Determine        |     | BMI category   |
|             |     | BMI category     |     |                |
+-------------+     +------------------+     +----------------+
```

## Comprehensive List of Python DataTypes

Below is a comprehensive list of data types in Python along with an example for each. You do not need to use, or even know all these datatypes. The list is provided for convenience and completeness.

3. **String (`str`)**:

   - Description: Sequence of characters enclosed in single, double, or triple quotes.

   - Example:

10. **Bytes (`bytes`)**:

    - Description: Immutable sequence of bytes.

    - Example:

````python
g = b'hello'
print(type(g))  

# Output: <class 'bytes'>
````

11. **Byte Array (`bytearray`)**:

    - Description: Mutable sequence of bytes.

    - Example:

````python
h = bytearray(5)
print(type(h))  

# Output: <class 'bytearray'>
````

12. **Complex (`complex`)**:

    - Description: Complex numbers, which have a real and imaginary part.

    - Example:

birth_date = date(int(birth_date_str[:4]), int(birth_date_str[5:7]), int(birth_date_str[8:10]))
current_date = date.today()
age = current_date.year - birth_date.year

# Adjust age if birthday hasn't occurred yet this year
birthday_this_year = date(current_date.year, birth_date.month, birth_date.day)
if birthday_this_year > current_date:
    age -= 1

# Calculate days until next birthday
if birthday_this_year < current_date:
    next_birthday = date(current_date.year + 1, birth_date.month, birth_date.day)
elif birthday_this_year == current_date:
    next_birthday = date(current_date.year + 1, birth_date.month, birth_date.day)
else:
    next_birthday = birthday_this_year

days_until_birthday = (next_birthday - current_date).days

# OUTPUT
print(f"Hello, {name}! You are {age} years old.")
if days_until_birthday == 0:
    print("Happy Birthday today!")
else:
    print(f"There are {days_until_birthday} days until your next birthday.")
