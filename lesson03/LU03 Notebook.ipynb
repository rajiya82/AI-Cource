# Lists

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

A **list** is an **ordered, mutable collection** of items. Lists can hold any type of data — integers, strings, floats, or even other lists.

Lists are defined using **square brackets** `[]`, and each item is separated by a comma.

Key characteristics:
- **Ordered**: Items maintain their position in the list
- **Mutable**: Items can be added, removed, or changed after creation
- **Zero-indexed**: The first item is at index `0`, the second at index `1`, and so on

## Defining a List

A list is created by assigning values to a variable using square brackets. Items can be variables or literal values.

## Accessing Elements

Use the index in square brackets to retrieve a specific item. Python supports **negative indexing**: `-1` refers to the last item, `-2` to the second-last, and so on.

## Modifying a List

Lists are mutable — you can change any item by assigning a new value to its index.

## Common List Methods

| Method | Description |
|---|---|
| `append(item)` | Adds an item to the end of the list |
| `insert(i, item)` | Inserts an item at position `i` |
| `remove(item)` | Removes the first occurrence of `item` |
| `pop(i)` | Removes and returns the item at index `i` (default: last item) |
| `clear()` | Removes all items from the list |
| `index(item)` | Returns the index of the first occurrence of `item` |
| `count(item)` | Returns how many times `item` appears |
| `sort()` | Sorts the list in ascending order (in place) |
| `reverse()` | Reverses the order of the list (in place) |
| `copy()` | Returns a shallow copy of the list |
| `extend(iterable)` | Appends all items from another iterable to the list |

## Membership Testing with `in`

The `in` operator checks whether a value exists in a list, returning `True` or `False`. Use it to test membership before accessing or removing an item:

```python
fruits = ["apple", "banana", "cherry"]

print("apple" in fruits)      # True
print("mango" in fruits)      # False
print("mango" not in fruits)  # True

# Practical use — check before acting:
if "banana" in fruits:
    fruits.remove("banana")
```

You will see `in` used in the scenario code that follows.


````python
# Defining a list using individual variables
item1 = "Item1"
item2 = "Item2"
item3 = "Item3"
list_name = [item1, item2, item3]
print(list_name)
# Output: ['Item1', 'Item2', 'Item3']

# Defining and initialising a list directly
fruits = ["apple", "banana", "cherry"]
print(fruits)
# Output: ['apple', 'banana', 'cherry']

# Accessing elements (zero-indexed)
print(fruits[0])   # First element
print(fruits[1])   # Second element
print(fruits[-1])  # Last element (negative indexing)
# Output:
# apple
# banana
# cherry

# Modifying a list item
fruits[2] = "blueberry"
print(fruits)
# Output: ['apple', 'banana', 'blueberry']
````

````python
# Demonstrating common list methods
my_list = [1, 2, 3, 4, 5]

# append() — add to the end
my_list.append(6)
print("append(6):", my_list)

# insert() — insert at a specific index
my_list.insert(0, 0)
print("insert(0, 0):", my_list)

# remove() — remove first occurrence of a value
my_list.remove(3)
print("remove(3):", my_list)

# pop() — remove and return item at index (default: last)
popped = my_list.pop()
print(f"pop(): removed {popped}, list is now {my_list}")

# clear() — remove all items
temp = my_list.copy()
temp.clear()
print("clear():", temp)

# index() — find position of a value
pos = my_list.index(4)
print("index(4):", pos)

# count() — count occurrences
my_list.append(2)
print("count(2):", my_list.count(2))

# sort() — sort in ascending order
temp = my_list.copy()
temp.sort()
print("sort():", temp)

# reverse() — reverse the list
temp = my_list.copy()
temp.reverse()
print("reverse():", temp)

# copy() — shallow copy
copied = my_list.copy()
print("copy():", copied)

# extend() — append all items from another iterable
my_list.extend([7, 8])
print("extend([7, 8]):", my_list)

# Output:
# append(6): [1, 2, 3, 4, 5, 6]
# insert(0, 0): [0, 1, 2, 3, 4, 5, 6]
# remove(3): [0, 1, 2, 4, 5, 6]
# pop(): removed 6, list is now [0, 1, 2, 4, 5]
# clear(): []
# index(4): 3
# count(2): 2
# sort(): [0, 1, 2, 2, 4, 5]
# reverse(): [5, 4, 2, 2, 1, 0]
# copy(): [0, 1, 2, 4, 5, 2]
# extend([7, 8]): [0, 1, 2, 4, 5, 2, 7, 8]
````

## Scenario: StockMaster Inc. — Product Inventory

StockMaster Inc. is a technology distributor that manages a warehouse of electronic products. The warehouse manager needs a simple system to:

1. **Store** the current product names in a list
2. **Access** individual products by their position in the list
3. **Add** newly arrived products
4. **Remove** discontinued products
5. **Check** whether a specific product is currently stocked
6. **Sort** the inventory alphabetically for a quick-reference report

The product inventory is stored as a **list of product names**. Each step below follows the **IPO (Input–Process–Output)** model introduced in Lesson 01.

## Code

````python
# StockMaster Inc. — Product Inventory (List)
# ================================================

# INPUT: build the initial inventory
inventory = ["Laptop", "Mouse", "Keyboard", "Monitor", "Webcam"]
print("Initial inventory:", inventory)
print("Total items in stock:", len(inventory))

# ACCESS: retrieve individual products by index
print("\nFirst product  (index  0):", inventory[0])
print("Second product (index  1):", inventory[1])
print("Last product   (index -1):", inventory[-1])

# PROCESS: add a new product arrival at the end
inventory.append("Headset")
print("\nAfter adding 'Headset':", inventory)

# PROCESS: insert a product at a specific position
inventory.insert(1, "USB Hub")   # insert after Laptop
print("After inserting 'USB Hub' at index 1:", inventory)

# PROCESS: remove a discontinued product
inventory.remove("Webcam")
print("\nAfter removing 'Webcam':", inventory)
print("Total items now:", len(inventory))

# PROCESS: check whether a product is in stock
product_to_find = "Keyboard"
if product_to_find in inventory:
    position = inventory.index(product_to_find)
    print(f"\n'{product_to_find}' is in stock at index {position}.")
else:
    print(f"\n'{product_to_find}' is NOT in stock.")

product_to_find = "Webcam"
if product_to_find in inventory:
    position = inventory.index(product_to_find)
    print(f"'{product_to_find}' is in stock at index {position}.")
else:
    print(f"'{product_to_find}' is NOT in stock.")

# OUTPUT: sort and display the final inventory
inventory.sort()
print("\nSorted inventory (A–Z):", inventory)

# Output:
# Initial inventory: ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam']
# Total items in stock: 5
# First product  (index  0): Laptop
# Second product (index  1): Mouse
# Last product   (index -1): Webcam
# After adding 'Headset': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Headset']
# After inserting 'USB Hub' at index 1: ['Laptop', 'USB Hub', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Headset']
# After removing 'Webcam': ['Laptop', 'USB Hub', 'Mouse', 'Keyboard', 'Monitor', 'Headset']
# Total items now: 6
# 'Keyboard' is in stock at index 3.
# 'Webcam' is NOT in stock.
# Sorted inventory (A–Z): ['Headset', 'Keyboard', 'Laptop', 'Monitor', 'Mouse', 'USB Hub']
````

# Dictionaries

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

A **dictionary** is an **unordered collection of key-value pairs**. Dictionaries are defined using **curly braces** `{}`, with each entry written as `key: value`.

Key characteristics:
- **Key-based access**: Values are retrieved by their key, not a numeric index
- **Mutable**: Entries can be added, updated, or removed after creation
- **Keys are unique**: Each key can appear only once in a dictionary
- **Keys must be immutable**: Strings, numbers, and tuples can be used as keys; lists cannot

Dictionaries are ideal for representing real-world entities where data has a meaningful label (e.g., a product's name, price, and stock level).

## Declaring a Dictionary

Use curly braces with `key: value` pairs separated by commas.

## Accessing Values

Use `dict["key"]` to access a value. Alternatively, use `dict.get("key")` which returns `None` (or a custom default) instead of raising a `KeyError` if the key is missing.

## Common Dictionary Methods

| Method | Description |
|---|---|
| `keys()` | Returns a view of all keys |
| `values()` | Returns a view of all values |
| `items()` | Returns a view of all key-value pairs |
| `get(key, default)` | Returns the value for `key`, or `default` if not found |
| `pop(key)` | Removes and returns the value for `key` |
| `popitem()` | Removes and returns the last inserted key-value pair |
| `update(dict2)` | Merges `dict2` into the dictionary |
| `setdefault(key, val)` | Returns value for `key`; inserts `key: val` if absent |
| `copy()` | Returns a shallow copy |
| `clear()` | Removes all entries |
| `fromkeys(keys, val)` | Creates a new dict with specified keys set to `val` |

````python
# Declaring a dictionary
student = {"name": "John", "age": 20, "course": "Computer Science"}
print(student)
# Output: {'name': 'John', 'age': 20, 'course': 'Computer Science'}

# Accessing values using keys
print(student["name"])   # Output: John
print(student["age"])    # Output: 20

# Safe access with .get() — returns None (or default) if key is missing
print(student.get("course"))             # Output: Computer Science
print(student.get("gpa", "Not set"))     # Output: Not set

# Adding and updating entries
student["gpa"] = 3.8
print(student)
# Output: {'name': 'John', 'age': 20, 'course': 'Computer Science', 'gpa': 3.8}

# Updating an existing entry
student["age"] = 21
print(student)

# Removing an entry
del student["gpa"]
print(student)
# Output: {'name': 'John', 'age': 21, 'course': 'Computer Science'}
````

````python
# Demonstrating common dictionary methods
my_dict = {"a": 1, "b": 2, "c": 3}

# keys(), values(), items()
print("keys():",   list(my_dict.keys()))
print("values():", list(my_dict.values()))
print("items():",  list(my_dict.items()))

# get() — safe retrieval with a default
print("get('a'):",    my_dict.get("a"))
print("get('z', 0):", my_dict.get("z", 0))

# pop() — remove a key and return its value
popped = my_dict.pop("a")
print(f"pop('a'): removed {popped}, dict is now {my_dict}")

# update() — merge another dictionary
my_dict.update({"d": 4, "e": 5})
print("update({'d': 4, 'e': 5}):", my_dict)

# setdefault() — insert key with default if not already present
val = my_dict.setdefault("f", 6)
print(f"setdefault('f', 6): returned {val}, dict is {my_dict}")

# copy()
copied = my_dict.copy()
print("copy():", copied)

# fromkeys() — create a dict from a list of keys
new_dict = dict.fromkeys(["x", "y", "z"], 0)
print("fromkeys(['x', 'y', 'z'], 0):", new_dict)

# clear()
temp = my_dict.copy()
temp.clear()
print("clear():", temp)
````

# Dictionaries — Scenario to Code

## Scenario: StockMaster Inc. — Product Record

The warehouse manager needs to store **detailed information** about each product — not just its name. A dictionary is the right tool because each piece of information has a meaningful label (key):

- `"name"` — product name
- `"category"` — product category
- `"price"` — selling price
- `"stock"` — units currently held in the warehouse

The first example below creates and manipulates a **single product record**. The second example shows how several product records are stored together in a **nested dictionary** (a dictionary where each value is itself a dictionary), using each product's SKU code as the key.

## Code

````python
# StockMaster Inc. — Single Product Record (Dictionary)
# ==========================================================

# INPUT: create the product record
product = {
    "name":     "Laptop",
    "category": "Electronics",
    "price":    1299.99,
    "stock":    42
}
print("Product record:", product)

# ACCESS: retrieve individual fields
print("\nProduct name:", product["name"])
print("Category:", product["category"])
print(f"Price:  ${product['price']:.2f}")
print("Units in stock:", product["stock"])

# PROCESS: update the price after a promotion
product["price"] = 1199.99
print(f"\nUpdated price: ${product['price']:.2f}")

# PROCESS: add a new field
product["sku"] = "LPT-001"
print("SKU added:", product["sku"])

# PROCESS: safe access — .get() returns a default instead of raising KeyError
discount = product.get("discount", 0)
print(f"\nDiscount (key missing — default used): {discount}%")
warranty = product.get("warranty", "Not specified")
print(f"Warranty: {warranty}")

# PROCESS: remove a field
del product["sku"]
print("\nAfter deleting 'sku':", product)

# OUTPUT: inspect all keys and values
print("\nAll keys:",   list(product.keys()))
print("All values:", list(product.values()))

# Output:
# Product record: {'name': 'Laptop', 'category': 'Electronics', 'price': 1299.99, 'stock': 42}
# Product name: Laptop
# Category: Electronics
# Price:  $1299.99
# Units in stock: 42
# Updated price: $1199.99
# SKU added: LPT-001
# Discount (key missing — default used): 0%
# Warranty: Not specified
# After deleting 'sku': {'name': 'Laptop', 'category': 'Electronics', 'price': 1199.99, 'stock': 42}
# All keys:   ['name', 'category', 'price', 'stock']
# All values: ['Laptop', 'Electronics', 1199.99, 42]
````

## Nested Dictionaries

A **nested dictionary** is a dictionary whose values are themselves dictionaries. This pattern is ideal when you need to store a **collection of records** where each record has multiple fields.

StockMaster stores its full product catalogue as a nested dictionary, keyed by each product's unique **SKU** (Stock Keeping Unit) code:

```python
catalogue = {
    "SKU-001": {"name": "Laptop",   "price": 1199.99, "stock": 37},
    "SKU-002": {"name": "Mouse",    "price":   29.99, "stock": 150},
    ...
}
```

To access a value inside a nested dictionary, chain two sets of square brackets:

```python
catalogue["SKU-001"]["price"]   # → 1199.99
catalogue["SKU-002"]["name"]    # → "Mouse"
```

````python
# StockMaster Inc. — Product Catalogue (Nested Dictionary)
# =============================================================

# INPUT: create the catalogue
catalogue = {
    "SKU-001": {"name": "Laptop",   "price": 1199.99, "stock": 37},
    "SKU-002": {"name": "Mouse",    "price":   29.99, "stock": 150},
    "SKU-003": {"name": "Keyboard", "price":   79.99, "stock": 80},
}

# ACCESS: retrieve a full product record
print("SKU-002 record:", catalogue["SKU-002"])

# ACCESS: retrieve individual fields from a nested record
print("Name:", catalogue["SKU-002"]["name"])
print(f"Price: ${catalogue['SKU-002']['price']:.2f}")
print("Stock:", catalogue["SKU-002"]["stock"])

# PROCESS: update stock for SKU-001 after selling 5 units
catalogue["SKU-001"]["stock"] = catalogue["SKU-001"]["stock"] - 5
print(f"\nSKU-001 stock after selling 5: {catalogue['SKU-001']['stock']}")

# PROCESS: add a new product to the catalogue
catalogue["SKU-004"] = {"name": "Webcam", "price": 59.99, "stock": 60}
print("Catalogue SKUs:", list(catalogue.keys()))

# PROCESS: check if a SKU exists using .get() with a default
sku_to_find = "SKU-005"
result = catalogue.get(sku_to_find, "Not found")
print(f"\nLooking up {sku_to_find}: {result}")

# PROCESS: update multiple fields at once using .update()
catalogue["SKU-003"].update({"price": 74.99, "on_sale": True})
print("SKU-003 after update:", catalogue["SKU-003"])

# OUTPUT: total number of products
print(f"\nTotal products in catalogue: {len(catalogue)}")

# Output:
# SKU-002 record: {'name': 'Mouse', 'price': 29.99, 'stock': 150}
# Name: Mouse
# Price: $29.99
# Stock: 150
# SKU-001 stock after selling 5: 32
# Catalogue SKUs: ['SKU-001', 'SKU-002', 'SKU-003', 'SKU-004']
# Looking up SKU-005: Not found
# SKU-003 after update: {'name': 'Keyboard', 'price': 74.99, 'stock': 80, 'on_sale': True}
# Total products in catalogue: 4
````

# Common Mistakes

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

Working with lists and dictionaries is fundamental, but there are several common pitfalls to be aware of. Understanding these mistakes — and how Python signals them — makes debugging much easier.

| Error Type | Typical Cause |
|---|---|
| **SyntaxError** | Using `()` instead of `[]` for a list, or missing a comma between items |
| **NameError** | Misspelling a variable name |
| **IndexError** | Accessing a list index that does not exist |
| **KeyError** | Accessing a dictionary key that has not been defined |
| **TypeError** | Using a mutable type (e.g. a list) as a dictionary key |

The code cell below demonstrates each mistake and shows how to handle or avoid it. `try/except` was introduced in Lesson 02 and is used here to catch each error gracefully.

````python
# COMMON MISTAKES WITH LISTS AND DICTIONARIES

# 1. Using () instead of [] creates a TUPLE, not a list
my_tuple = (1, 2, 3)   # tuple — immutable, cannot be modified
my_list  = [1, 2, 3]   # list  — mutable
print(f"Parentheses give a: {type(my_tuple)}")    # <class 'tuple'>
print(f"Square brackets give a: {type(my_list)}") # <class 'list'>

# 2. NameError — misspelling a variable name
fruits = ["apple", "banana", "cherry"]
try:
    print(fruit)   # 'fruit' is not defined (missing the 's')
except NameError as e:
    print(f"NameError: {e}")

# 3. IndexError — accessing an index beyond the list length
fruits = ["apple", "banana", "cherry"]   # valid indices: 0, 1, 2
try:
    print(fruits[3])   # index 3 does not exist in a 3-item list
except IndexError as e:
    print(f"IndexError: {e}")
# Fix: check length before accessing
print(f"List has {len(fruits)} items. Valid indices: 0 to {len(fruits) - 1}")

# 4. KeyError — accessing a key that was never added
book = {"title": "1984", "author": "George Orwell"}
try:
    print(book["publish_date"])   # 'publish_date' key does not exist
except KeyError as e:
    print(f"KeyError: {e}")
# Fix: use .get() with a default value
publish_date = book.get("publish_date", "Unknown")
print(f"Publish date (safe access): {publish_date}")

# 5. TypeError — using a mutable type (list) as a dictionary key
try:
    invalid_dict = {["apple", "banana"]: "fruits"}   # lists are not hashable
except TypeError as e:
    print(f"TypeError: {e}")
# Fix: use a tuple (immutable) as the key instead
valid_dict = {("apple", "banana"): "fruits"}
print(f"Using tuple as key: {valid_dict}")

# Output:
# Parentheses give a: <class 'tuple'>
# Square brackets give a: <class 'list'>
# NameError: name 'fruit' is not defined
# IndexError: list index out of range
# List has 3 items. Valid indices: 0 to 2
# KeyError: 'publish_date'
# Publish date (safe access): Unknown
# TypeError: unhashable type: 'list'
# Using tuple as key: {('apple', 'banana'): 'fruits'}
````

**OUTPUT**
