# Introduction to SQL and Relational Databases

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

**SQL** (Structured Query Language) is the standard language for relational database management systems (RDBMS). It was developed in the 1970s by IBM researchers Raymond Boyce and Donald Chamberlin. SQL has become the fundamental language for creating, managing, and querying databases in both commercial and open-source systems such as MySQL, PostgreSQL, Microsoft SQL Server, Oracle Database, and SQLite.

## What is a Relational Database?

A relational database organises data into **tables** with rows and columns. Each row represents a record and each column represents an attribute of that record. Relationships between tables are established using keys. The model was proposed by E. F. Codd in 1970 and emphasises data accuracy, integrity, and the ability to support complex queries and transactions.

## Core SQL Concepts

| Concept | Description |
|---|---|
| **DDL** — Data Definition Language | Commands for defining the schema: `CREATE`, `ALTER`, `DROP` |
| **DML** — Data Manipulation Language | Commands for inserting, updating, and deleting data: `INSERT`, `UPDATE`, `DELETE` |
| **DQL** — Data Query Language | Commands for querying and retrieving data: `SELECT` |
| **DCL** — Data Control Language | Commands for controlling access to data: `GRANT`, `REVOKE` |
| **Transactions** | Ensures data integrity through atomic and consistent operations |
| **Primary Key** | A column (or combination) that uniquely identifies each row in a table |
| **Foreign Key** | A column that creates a link between two tables by referencing a primary key |
| **Normalisation** | Process of organising data to eliminate redundancy and dependency |

## Usage and Applications

Relational databases are used across many fields:

- **Business and Finance**: customer data, financial transactions, inventory
- **Healthcare**: patient records, billing, medical research
- **Education**: student records, course registrations, grading
- **E-commerce**: user accounts, product catalogues, orders
- **Government**: public records, legislation, public services data

# SQLite

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

**SQLite** is a lightweight, fast, and self-contained SQL database engine. Unlike most other database systems, SQLite does not require a separate server process — the entire database is stored as a **single file** on disk. This makes it exceptionally portable and easy to deploy.

SQLite is one of the most widely used database systems in the world, found in mobile devices, desktop applications, browsers, and embedded systems.

## Key Features

| Feature | Description |
|---|---|
| **Single file** | The entire database is one cross-platform file — easy to copy and share |
| **ACID compliance** | Transactions comply with Atomicity, Consistency, Isolation, Durability |
| **Dynamic typing** | Data types are associated with individual values, not columns |
| **Lightweight** | Small footprint — ideal for resource-constrained or embedded applications |
| **Serverless** | No server setup, configuration, or administration required |
| **Cross-platform** | Works on Windows, Linux, and macOS |

## Installing SQLite (Windows)

## Connecting to SQLite from Python

Python includes the `sqlite3` module in its standard library — no installation required.

| Method / Object | Description |
|---|---|
| `sqlite3.connect('file.db')` | Opens an existing database or creates a new one |
| `conn.cursor()` | Creates a **cursor** — an object used to execute SQL commands and fetch results |
| `conn.commit()` | Saves (commits) any pending changes to the database |
| `conn.close()` | Closes the database connection and frees resources |

Think of the **cursor** as a messenger: you send it a SQL request, it executes the command against the database, and returns the results to you.

## Creating Database Tables

First we will create the tables. If the tables already exist they will be dropped first, then recreated. Doing so enables us to re-start our code with a fresh database whenever we want.

```python
import sqlite3

# Connect to the database and create a cursor
conn = sqlite3.connect('school.db')
c = conn.cursor()

# Drop the existing tables if they exist
c.execute('DROP TABLE IF EXISTS enrollments')
c.execute('DROP TABLE IF EXISTS students')
c.execute('DROP TABLE IF EXISTS courses')

# Create the new tables
c.execute('''
          CREATE TABLE students
          (id INTEGER PRIMARY KEY AUTOINCREMENT,
          name TEXT NOT NULL,
          age INTEGER NOT NULL,
          grade TEXT NOT NULL)
          ''')

c.execute('''
          CREATE TABLE courses
          (id INTEGER PRIMARY KEY AUTOINCREMENT,
          name TEXT NOT NULL,
          teacher TEXT NOT NULL)
          ''')

c.execute('''
          CREATE TABLE enrollments
          (student_id INTEGER NOT NULL,
          course_id INTEGER NOT NULL,
          FOREIGN KEY (student_id) REFERENCES students(id),
          FOREIGN KEY (course_id) REFERENCES courses(id),
          PRIMARY KEY (student_id, course_id))
          ''')

# Print the contents of the three tables, which will be blank.
print(c.execute('SELECT * FROM students').fetchall())
print(c.execute('SELECT * FROM courses').fetchall())
print(c.execute('SELECT * FROM enrollments').fetchall())

# Commit the changes and close the connection
conn.commit()
c.close()
conn.close()

# Output:
# []   << Indicates that the students table is empty
# []   << Indicates that the courses table is empty
# []   << Indicates that the enrollments table is empty
```

## Inserting Data

Next, we will insert some data into the tables. Note that if you re-run this cell it will **first delete all student and course data**. You can do this whenever you wish to get a 'fresh' database.

Two methods used in the code below are worth noting:

| Method / Function | Description |
|---|---|
| `c.executemany(sql, data)` | Executes the same SQL statement repeatedly for each item in `data` — more efficient than calling `execute()` once per row |
| `pd.read_sql_query(sql, conn)` | Runs a SQL query and returns the result directly as a Pandas DataFrame |

Student data is printed to show you how the output looks. Notice how we also use the `pandas` library to nicely format our results using `display` rather than `print` for both the student and course data.

```python
import pandas as pd
import sqlite3

# Connect to the database and create a cursor
conn = sqlite3.connect('school.db')
c = conn.cursor()

# Remove all data from students, courses, and enrollments tables
c.execute('DELETE FROM enrollments')
c.execute('DELETE FROM students')
c.execute('DELETE FROM courses')

# Insert data
students_data = [(1, 'Tan Ah Kow', 20, 'A'),
                 (2, 'Lim Siew Kim', 22, 'B'),
                 (3, 'Wong Meng Choy', 21, 'B'),
                 (4, 'Nurul Ayuni', 22, 'A'),
                 (5, 'Rajesh Kumar', 20, 'C'),
                 (6, 'Lee Hui Ying', 23, 'B'),
                 (7, 'Chen Wei Ming', 22, 'A'),
                 (8, 'Siti Aishah', 21, 'B'),
                 (9, 'Mohammed Faizal', 20, 'C'),
                 (10, 'Goh Ah Lek', 24, 'C')]

courses_data = [(1, 'Math', 'Mr. A'),
                (2, 'Science', 'Ms. B'),
                (3, 'English', 'Mr. C'),
                (4, 'History', 'Ms. D'),
                (5, 'Geography', 'Mr. E')]

enrollments_data = [(1, 1), (1, 2), (1, 3),
                    (2, 1), (2, 2), (2, 3),
                    (3, 1), (3, 2), (3, 3),
                    (4, 1), (4, 2), (4, 3),
                    (5, 1), (5, 2), (5, 3),
                    (6, 1), (6, 2), (6, 3),
                    (7, 1), (7, 2), (7, 3),
                    (8, 1), (8, 2), (8, 3),
                    (9, 1), (9, 2), (9, 3),
                    (10, 1), (10, 2), (10, 3)]

c.executemany('INSERT INTO students VALUES (?, ?, ?, ?)', students_data)
c.executemany('INSERT INTO courses VALUES (?, ?, ?)', courses_data)
c.executemany('INSERT INTO enrollments (student_id, course_id) VALUES (?, ?)', enrollments_data)

# Save (commit) the changes
conn.commit()

# Uncomment the following line to see how normal print() displays data
c.execute('SELECT * FROM students LIMIT 5')
print(c.fetchall())

# Print the first 5 rows of the students table
df_students = pd.read_sql_query('SELECT * FROM students LIMIT 5', conn)
display(df_students)

# Print the first 5 rows of the courses table
df_courses = pd.read_sql_query('SELECT * FROM courses LIMIT 5', conn)
display(df_courses)

# Print the first 5 rows of the enrollments table
df_enrollments = pd.read_sql_query('SELECT * FROM enrollments LIMIT 5', conn)
display(df_enrollments)

# Close the cursor and the connection
c.close()
conn.close()
```

We have already seen some basic examples of inserting data, retrieving data and deleting data using SQL in Python. In this section we will look at these operations in more detail. 

The examples below use `pd.read_sql_query(sql, conn)` to run a SQL query and return the result directly as a Pandas DataFrame — this is the most convenient way to display and work with query results in Jupyter.

## SELECT

The `SELECT` statement retrieves data from one or more tables. `SELECT *` returns all columns; you can also name specific columns.

### Simple SELECT

Connects to the database and retrieves the first five rows from the `students` and `courses` tables using `SELECT * FROM table LIMIT 5`.

```python
import sqlite3
import pandas as pd

# Connect to the database and create a cursor
conn = sqlite3.connect('school.db')

# Print the first 5 rows of the students table
df_students = pd.read_sql_query('SELECT * FROM students LIMIT 5', conn)
display(df_students)

# Print the first 5 rows of the courses table
df_courses = pd.read_sql_query('SELECT * FROM courses LIMIT 5', conn)
display(df_courses)

# Close the connection
conn.close()
```

### SELECT with a Filter

A `WHERE` clause filters rows by a condition. Here we retrieve only students whose grade is `'A'`.

```python
import sqlite3
import pandas as pd

# Connect to the database and create a cursor
conn = sqlite3.connect('school.db')

# Print the students who are in grade 'A'
df_students_a_grade = pd.read_sql_query("SELECT * FROM students WHERE grade = 'A'", conn)
display(df_students_a_grade)

# Close the connection
conn.close()
```

### SELECT with a Join

A `JOIN` combines rows from two or more tables based on a related column. This query joins `enrollments`, `students`, and `courses` to display each student's name alongside the courses they are enrolled in.

- `JOIN students ON students.id = enrollments.student_id` — matches each enrolment to a student
- `JOIN courses ON courses.id = enrollments.course_id` — matches each enrolment to a course

```python
import sqlite3
import pandas as pd

# Connect to the database and create a cursor
conn = sqlite3.connect('school.db')

# Print the student names along with the courses they are enrolled in
df_enrollments = pd.read_sql_query('''
    SELECT students.name, courses.name AS course_name
    FROM enrollments
    JOIN students ON students.id = enrollments.student_id
    JOIN courses ON courses.id = enrollments.course_id
    ''', conn)
display(df_enrollments)

# Close the connection
conn.close()
```

### SELECT with a Join and a Filter

Combines a `JOIN` with a `WHERE` clause to narrow the results further. Here we retrieve only the students in grade `'A'` and the courses they are enrolled in.

```python
import sqlite3
import pandas as pd

# Connect to the database and create a cursor
conn = sqlite3.connect('school.db')

# Print the student names along with the courses they are enrolled in, filter for grade 'A'
df_enrollments_a_grade = pd.read_sql_query('''
    SELECT students.name, courses.name AS course_name
    FROM enrollments
    JOIN students ON students.id = enrollments.student_id
    JOIN courses ON courses.id = enrollments.course_id
    WHERE students.grade = 'A'
    ''', conn)
display(df_enrollments_a_grade)

# Close the connection
conn.close()
```

## UPDATE

The UPDATE statement is used to update existing records in a table.

```python
import sqlite3
import pandas as pd

# Connect to the database
conn = sqlite3.connect('school.db')
c = conn.cursor()

# Print the first 5 rows of the students table show data BEFORE the update
df_students = pd.read_sql_query('SELECT * FROM students LIMIT 5', conn)
display(df_students)

# Update the age of 'Tan Ah Kow' to 21
c.execute("UPDATE students SET age = 21 WHERE name = 'Tan Ah Kow'")

# Commit the changes
conn.commit()

# Print the first 5 rows of the students table to verify the update
df_students = pd.read_sql_query('SELECT * FROM students LIMIT 5', conn)
display(df_students)

# Close the cursor and the connection
c.close()
conn.close()
```

## DELETE

The DELETE statement is used to remove existing records in a table.

Note 1: If you want to re-run the code in this notebook, you will need to re-run the code in the first cell to create the tables and insert the data.

Note 2: If you want to remove all records from a table, you can use the DELETE statement without a WHERE clause.

```python
import sqlite3
import pandas as pd

# Connect to the database
conn = sqlite3.connect('school.db')
c = conn.cursor()

# Print the first 5 rows of the students table show data BEFORE the deletion
df_students = pd.read_sql_query('SELECT * FROM students LIMIT 5', conn)
display(df_students)

# Delete the student 'Tan Ah Kow'
c.execute("DELETE FROM students WHERE name = 'Tan Ah Kow'")

# Commit the changes
conn.commit()

# Print the first 5 rows of the students table to verify the deletion
df_students = pd.read_sql_query('SELECT * FROM students LIMIT 5', conn)
display(df_students)

# Close the cursor and the connection
c.close()
conn.close()
```
