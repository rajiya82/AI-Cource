# Nested If Statements

<a id="Nested-If-Statements"></a>

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

A **nested if statement** is an `if` statement placed inside another `if` statement. This allows your program to check a second condition only after a first condition has already been established as true.

The general structure is:

```python
if outer_condition:
    if inner_condition:
        # both outer and inner are true
    else:
        # outer is true, but inner is false
else:
    # outer condition is false
```

**When to use nested if statements:** whenever a decision depends on more than one condition at the same time. For example: "If it is a weekday pass AND the purchaser is a child, the price is $45." You cannot determine the price from the pass type alone — you also need to know the age group.

**Indentation is critical.** Each level of nesting adds four more spaces of indentation. Python uses indentation to decide which block a line belongs to. A misplaced line will produce wrong results without giving you any error message.

## A Simple Example — Cinema Ticket Pricing

A cinema charges different prices depending on session time and customer age:

- **Morning session** (`m`): Child (under 13) → $5 &nbsp;|&nbsp; Everyone else → $8
- **Other sessions**: Child → $7 &nbsp;|&nbsp; Everyone else → $12

The outer `if` checks the **session time**. Only after knowing the session do we check the **age**. Notice how the indentation clearly shows which code belongs to which branch.

````python
# Cinema ticket pricing — two-level nested if statement
time_of_day = input("Enter session (m = morning, a = afternoon, e = evening): ").lower()
age = int(input("Enter your age: "))

if time_of_day == 'm':
    if age < 13:
        price = 5
    else:
        price = 8
else:
    if age < 13:
        price = 7
    else:
        price = 12

print("Ticket price: $" + str(price))

# Sample outputs:
# Session m, Age 10  ->  Ticket price: $5
# Session m, Age 25  ->  Ticket price: $8
# Session e, Age 10  ->  Ticket price: $7
# Session e, Age 30  ->  Ticket price: $12
````

## A More Complex Example — Student Grading

The example below adds a second variable: first it checks the exam total (60 or 100 marks), then it checks the student's score within that scale. Each branch of the outer `if` contains its own complete `if-elif-else` chain.

Study the indentation carefully. Each level of nesting is indented four more spaces relative to the one above it. The `elif` keyword (short for "else if") lets you chain multiple conditions within a single branch without needing additional nesting.

````python
# Student grading — nested if with elif chains
total_mark = input("What is the total mark (60 or 100): ")
total_mark = int(total_mark)

mark = input(f"Enter your mark (0-{total_mark}): ")
mark = int(mark)

if total_mark == 100:
    if mark >= 70:
        grade = 'A'
    elif mark >= 60:
        grade = 'B'
    elif mark >= 50:
        grade = 'C'
    else:
        grade = 'F'
else:
    if mark >= 50:
        grade = 'A'
    elif mark >= 40:
        grade = 'B'
    elif mark >= 30:
        grade = 'C'
    else:
        grade = 'F'

print("Your Grade is " + grade)
````
