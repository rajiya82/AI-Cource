# The `while` Loop

<a id="The-while-Loop"></a>

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

The `while` loop repeatedly executes a block of code **as long as a condition remains true**. It is used when the number of iterations is not known in advance.

| Component | Description |
|---|---|
| `while` | Keyword that begins the loop |
| `condition` | Evaluated before each iteration — loop runs while `True` |
| `:` | Marks the start of the loop body |
| `statement(s)` | Indented code that executes on each iteration |

```python
while condition:
    # statement(s)
```

## Example 1: Counting from 1 to 5

Initialise a counter before the loop. Increment it inside the loop. The loop ends when the condition becomes `False`.

````python
count = 1
while count <= 5:
    print(count)
    count += 1
````

## Example 2: Repeating Until a Condition Is Met

Use `while True` with a `break` when the exit condition is checked **after** reading input. This avoids initialising a dummy value before the loop.

````python
while True:
    user_input = input("Enter 'quit' to exit: ")
    if user_input == "quit":
        break
    print("You entered:", user_input)
print("Goodbye!")
````

## Example 3: Data Validation

Keep prompting until the user provides a valid value. Here the program rejects passwords shorter than 8 characters.

````python
password = input("Enter a password: ")
while len(password) < 8:
    print("Password must be at least 8 characters long.")
    password = input("Try again: ")
print("Password set successfully!")
````

## `break` and `continue`

| Statement | Effect |
|---|---|
| `break` | Exits the loop immediately |
| `continue` | Skips the rest of the current iteration and moves to the next |

**`break`** — stop as soon as a negative number is entered:

````python
while True:
    number = int(input("Enter a positive number (negative to quit): "))
    if number < 0:
        print("Negative number detected. Exiting.")
        break
    print(f"You entered: {number}")
````

**`continue`** — skip even numbers, print only odd numbers from 1 to 10:

````python
count = 0
while count < 10:
    count += 1
    if count % 2 == 0:   # even — skip
        continue
    print(count)          # only odd numbers reach this line
````

# The `for` Loop

<a id="The-for-Loop"></a>

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

The `for` loop iterates over a sequence and executes the loop body once per item. It is used when the number of iterations is **known in advance**.

| Component | Description |
|---|---|
| `for` | Keyword that begins the loop |
| `variable` | Takes the value of each item per iteration |
| `in` | Specifies the sequence to iterate over |
| `iterable` | Any sequence: `range()`, list, string, etc. |
| `:` | Marks the start of the loop body |

```python
for variable in iterable:
    # statement(s)
```

## The `range()` Function

`range()` generates a sequence of integers. It is the most common way to drive a `for` loop when you need a specific number of iterations.

| Usage | Generated sequence |
|---|---|
| `range(5)` | `0, 1, 2, 3, 4` |
| `range(1, 6)` | `1, 2, 3, 4, 5` |
| `range(0, 10, 2)` | `0, 2, 4, 6, 8` (step of 2) |
| `range(5, 0, -1)` | `5, 4, 3, 2, 1` (counting down) |

## Example 1: Counting from 1 to 5

`range(1, 6)` generates the integers 1 through 5. Compare this with the `while` loop version in Example 1 above — the `for` loop is more concise.

````python
for number in range(1, 6):
    print(number)
````

## Example 2: Counting Down

`range()` accepts a negative step to count backwards. This is a common pattern for countdowns.

````python
for number in range(5, 0, -1):
    print(number)
print("Blast off!")
````

## Example 3: Accumulating a Sum

The **accumulator pattern**: initialise a variable to `0`, then add to it on each iteration.

> **Note:** Avoid naming variables `sum`, `list`, `input`, or `print` — > these are Python built-in functions and overwriting them causes hard-to-find bugs.

````python
total = 0
for i in range(1, 6):   # 1 + 2 + 3 + 4 + 5
    total += i
print("Sum:", total)    # Sum: 15
````

## The Underscore (`_`) as a Throwaway Variable

By convention, `_` signals that the loop variable is **intentionally not used** inside the loop body. Use it when you only need to repeat an action a fixed number of times.

```python
# Print "Hello!" five times — the counter value is irrelevant
for _ in range(5):
    print("Hello!")
```

# Lists (Revisit from Lesson 03)

<a id="Lists"></a>

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

A **list** is an ordered, **mutable** (changeable) collection of items enclosed in square brackets `[]`. Lists can hold values of any type — integers, strings, or even other lists.

| Operation | Syntax | Result |
|---|---|---|
| Create empty | `my_list = []` | `[]` |
| Create with values | `my_list = [1, 2, 3]` | `[1, 2, 3]` |
| Add element | `my_list.append(4)` | `[1, 2, 3, 4]` |
| Access by index | `my_list[0]` | `1` (zero-indexed) |
| Get length | `len(my_list)` | `4` |
| Check membership | `2 in my_list` | `True` |

````python
fruits = ["apple", "banana", "cherry"]

print(fruits)           # ["apple", "banana", "cherry"]
print(fruits[0])        # apple   (first element)
print(fruits[-1])       # cherry  (last element)
print(len(fruits))      # 3

fruits.append("date")
print(fruits)           # ["apple", "banana", "cherry", "date"]
````

# Looping over Lists

<a id="Looping-over-Lists"></a>

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

The `for` loop can iterate directly over a list — no index required. On each iteration, the loop variable takes the value of the next element. This is commonly called the **for-each** pattern.

```python
for element in my_list:
    # statement(s)
```

## Example 1: Greeting Each Name in a List

The loop variable `name` takes the value `"Alice"`, then `"Bob"`, then `"Charlie"` on successive iterations.

````python
names = ["Alice", "Bob", "Charlie"]
for name in names:
    print("Hello,", name)
````

## Example 2: Calculating the Total Length of Words

`len()` returns the number of characters in a string. Combine it with the accumulator pattern to total the lengths of all words in a list.

````python
words = ["apple", "banana", "cherry"]
total_length = 0
for word in words:
    total_length += len(word)
print("Total length:", total_length)   # 5 + 6 + 6 = 17
````

## Example 3: Summing a List of Numbers

The same accumulator pattern works with a list of numbers. Notice the loop variable `num` takes each value from the list directly — no indexing needed.

````python
numbers = [10, 20, 30, 40, 50]
total = 0
for num in numbers:
    total += num
print("Sum:", total)   # 150
````

# Useful Built-in Functions

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

Python provides several built-in functions that work directly with lists and sequences. These save you from writing loops for common operations and make your code more readable.

| Function | What it does | Example |
|---|---|---|
| `min(iterable)` | Returns the smallest value | `min([3, 1, 4])` → `1` |
| `max(iterable)` | Returns the largest value | `max([3, 1, 4])` → `4` |
| `sum(iterable)` | Returns the total of all values | `sum([3, 1, 4])` → `8` |
| `sorted(iterable)` | Returns a new sorted list (original unchanged) | `sorted([3, 1, 4])` → `[1, 3, 4]` |
| `enumerate(iterable)` | Returns index-value pairs as you loop | `for i, v in enumerate(["a","b"])` → `(0,"a"), (1,"b")` |

> **Note**: You used the accumulator pattern to calculate a total in the previous section. `sum()` does that in one line. Similarly, instead of writing a loop to find the highest value, `max()` gives you the answer directly. You will see all of these used in Lesson 07 (NumPy and Pandas) where they are fundamental operations on data.

````python
scores = [85, 42, 91, 67, 78, 55]

# min() and max() — smallest and largest values
print(f"Lowest score:  {min(scores)}")
print(f"Highest score: {max(scores)}")

# sum() — total of all values
total = sum(scores)
average = total / len(scores)
print(f"Total:   {total}")
print(f"Average: {average:.1f}")

# sorted() — returns a new sorted list; original is unchanged
ranked = sorted(scores)
print(f"Original: {scores}")
print(f"Sorted:   {ranked}")
print(f"Reverse:  {sorted(scores, reverse=True)}")

# enumerate() — gives you the index alongside each value
print("\nScore report:")
for position, score in enumerate(sorted(scores, reverse=True), start=1):
    print(f"  Rank {position}: {score}")

# Output:
# Lowest score:  42
# Highest score: 91
# Total:   418
# Average: 69.7
# Original: [85, 42, 91, 67, 78, 55]
# Sorted:   [42, 55, 67, 78, 85, 91]
# Reverse:  [91, 85, 78, 67, 55, 42]
#
# Score report:
#   Rank 1: 91
#   Rank 2: 85
#   Rank 3: 78
#   Rank 4: 67
#   Rank 5: 55
#   Rank 6: 42
````

# Tuples

<a id="Tuples"></a>

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

A **tuple** is an ordered, **immutable** (unchangeable) collection of items enclosed in parentheses `()`. Once created, its values cannot be modified. Tuples are used to group related values that belong together and should not change.

| Feature | List | Tuple |
|---|---|---|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` |
| Mutable | Yes | **No** |
| Use when | Collection may change | Values are fixed |

````python
# Creating tuples
point = (3, 7)             # a 2D coordinate
rgb   = (255, 128, 0)      # an RGB colour

# Tuple unpacking — assign multiple variables at once
x, y = point
print(f"x = {x}, y = {y}")   # x = 3, y = 7

# Parentheses are optional
a, b = 0, 1
print(a, b)               # 0 1

# Swapping two variables using tuple unpacking
a, b = b, a
print(a, b)               # 1 0
````

## Iterating over a List of Tuples

When a list contains tuples, you can **unpack** each tuple directly in the `for` statement.

````python
coordinates = [(1, 2), (3, 4), (5, 6)]
for x, y in coordinates:
    print(f"x = {x}, y = {y}")
````

# Dictionaries (Revisit from Lesson 03)

<a id="Dictionaries"></a>

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

A **dictionary** is an unordered, **mutable** collection of **key-value pairs** enclosed in curly braces `{}`. Each key is unique and is used to look up its associated value — like a word and its definition.

| Operation | Syntax | Result |
|---|---|---|
| Create empty | `my_dict = {}` | `{}` |
| Create with values | `{"a": 1, "b": 2}` | `{"a": 1, "b": 2}` |
| Add or update | `my_dict["key"] = value` | — |
| Read value | `my_dict["key"]` | the value |
| Check for key | `"key" in my_dict` | `True` or `False` |
| Get all keys | `my_dict.keys()` | — |
| Get all values | `my_dict.values()` | — |

````python
# A product represented as a dictionary
product = {"Name": "Socks", "ID": "1", "Price": 5.0, "Quantity": 10}

print(product["Name"])       # Socks
print(product["Price"])      # 5.0

# Add a new key
product["Colour"] = "Blue"
print(product)

# Check if a key exists
print("ID" in product)       # True
print("Weight" in product)   # False
````

# Looping over Dictionaries

<a id="Looping-over-Dictionaries"></a>

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

Dictionaries provide three methods for iteration:

| Method | Iterates over | Example |
|---|---|---|
| `.keys()` | Keys only | `for k in d.keys():` |
| `.values()` | Values only | `for v in d.values():` |
| `.items()` | Key-value pairs (as tuples) | `for k, v in d.items():` |

## Example 1: Iterating over Values

Use `.values()` when you only need the values, not the keys. Here we total the prices of all items.

````python
prices = {"apple": 1.20, "banana": 0.50, "cherry": 2.00}

total = 0
for value in prices.values():
    total += value
print(f"Total cost: ${total:.2f}")   # $3.70
````

## Example 2: Iterating over Key-Value Pairs

Use `.items()` when you need both the key and the value. The loop unpacks each key-value pair into two variables.

````python
prices = {"apple": 1.20, "banana": 0.50, "cherry": 2.00}

for item, price in prices.items():
    print(f"  {item}: ${price:.2f}")
````

# Lists of Dictionaries

<a id="Lists-of-Dictionaries"></a>

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

A **list of dictionaries** combines both structures: the list holds multiple records, and each record is a dictionary. This is the natural way to represent a collection of structured items such as products in an inventory.

| Operation | Syntax |
|---|---|
| Create | `inventory = []` |
| Add a record | `inventory.append({"Name": "Socks", ...})` |
| Access a record | `inventory[0]` |
| Access a field | `inventory[0]["Name"]` |

````python
inventory = [
    {"Name": "Socks",  "ID": "1", "Price":  5.0, "Quantity": 10},
    {"Name": "Shoes",  "ID": "2", "Price": 20.0, "Quantity":  5},
    {"Name": "Gloves", "ID": "3", "Price":  8.0, "Quantity": 20},
]

# Access a specific field
print(inventory[1]["Name"])    # Shoes

# Loop over all records
for item in inventory:
    print(f"{item['Name']} | ${item['Price']:.2f} | Qty: {item['Quantity']}")
````

## Searching within a List of Dictionaries

A common pattern is to loop over the list and check a specific field to find a matching record.

````python
inventory = [
    {"Name": "Socks",  "ID": "1", "Price":  5.0, "Quantity": 10},
    {"Name": "Shoes",  "ID": "2", "Price": 20.0, "Quantity":  5},
]

# Find a product by ID
search_id = "2"
for item in inventory:
    if item["ID"] == search_id:
        print(f"Found: {item['Name']} at ${item['Price']:.2f}")
        break
````

# Nested Loops

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

A **nested loop** is a loop placed inside another loop. The **inner loop** runs completely for **every single iteration** of the outer loop.

```python
for outer_item in outer_sequence:
    for inner_item in inner_sequence:
        # runs for every (outer_item, inner_item) combination
```

Nested loops are most commonly used when working with **lists of dictionaries** — for example, when processing a shopping cart against a product inventory. The outer loop goes through each cart item, and the inner loop searches the inventory for a matching product ID. A `break` exits the inner loop as soon as a match is found so the search does not continue unnecessarily.

## Example: Match Cart Items Against an Inventory

````python
# Nested loop: match each cart item against the inventory

cart = [
    {"ID": "2", "Quantity": 1},
    {"ID": "3", "Quantity": 2},
]
demo_inventory = [
    {"Name": "Socks",  "ID": "1", "Price":  5.0},
    {"Name": "Shoes",  "ID": "2", "Price": 20.0},
    {"Name": "Gloves", "ID": "3", "Price":  8.0},
]

# Outer loop: iterate over every item in the cart
for cart_item in cart:
    # Inner loop: search for a matching product in the inventory
    for product in demo_inventory:
        if product["ID"] == cart_item["ID"]:
            qty  = cart_item["Quantity"]
            cost = product["Price"] * qty
            print(f"  {product['Name']} x{qty} = ${cost:.2f}")
            break   # match found — stop searching the inventory

# Output:
#   Shoes x1 = $20.00
#   Gloves x2 = $16.00
````

# List Comprehensions

<a href="#Table-of-Contents"><button style="background-color:#0072CE;color:white;border:none;padding:5px 14px;border-radius:5px;cursor:pointer;font-size:0.85em;">&#8593; Back to Index</button></a>

A **list comprehension** is a concise, readable way to build a new list from an existing iterable. Instead of initialising an empty list and calling `append()` inside a `for` loop, the entire operation fits on one line.

**Basic syntax:**
```python
new_list = [expression for item in iterable]
```

**Syntax with a condition (filter):**
```python
new_list = [expression for item in iterable if condition]
```

**Comparison — `for` loop vs list comprehension:**
```python
# Using a for loop
squares = []
for x in range(1, 6):
    squares.append(x ** 2)

# Using a list comprehension — identical result, one line
squares = [x ** 2 for x in range(1, 6)]
# → [1, 4, 9, 16, 25]
```

> **Rule of thumb:** Use a list comprehension when you are *creating a new list*. Use a regular `for` loop when you are *modifying an existing structure* (e.g. updating inventory quantities) — the intent is clearer in those cases.

````python
# Basic list comprehensions

# 1. Squares of 1 to 5
squares = [x ** 2 for x in range(1, 6)]
print("Squares:", squares)
# Output: [1, 4, 9, 16, 25]

# 2. Double each value
numbers = [1, 2, 3, 4, 5]
doubled = [n * 2 for n in numbers]
print("Doubled:", doubled)
# Output: [2, 4, 6, 8, 10]

# 3. Apply a 10% discount to a list of prices
prices = [100.0, 49.99, 250.0, 15.0]
discounted = [round(p * 0.9, 2) for p in prices]
print("Discounted:", discounted)
# Output: [90.0, 44.99, 225.0, 13.5]
````

## List Comprehensions with a Condition

Adding an `if` clause **filters** the source — only items for which the condition is `True` are included in the new list:

```python
new_list = [expression for item in iterable if condition]
```

````python
# List comprehensions with a condition (filter)

# 1. Only even numbers from 1 to 10
evens = [x for x in range(1, 11) if x % 2 == 0]
print("Evens:", evens)
# Output: [2, 4, 6, 8, 10]

# 2. Numbers greater than 5
numbers = [3, 7, 1, 9, 4, 6, 2, 8]
big = [n for n in numbers if n > 5]
print("Greater than 5:", big)
# Output: [7, 9, 6, 8]

# 3. Prices above $50 only
prices = [100.0, 49.99, 250.0, 15.0, 75.0]
expensive = [p for p in prices if p > 50]
print("Prices above $50:", expensive)
# Output: [100.0, 250.0, 75.0]
````

## List Comprehensions with Lists of Dictionaries

List comprehensions work especially well for **extracting a single field** from every record in a list of dictionaries, turning structured data into a plain list:

```python
# Extract the "Name" field from every product
names = [item["Name"] for item in inventory]
```

Adding a condition filters the records first:

```python
# Only names of products that are currently in stock
in_stock = [item["Name"] for item in inventory if item["Quantity"] > 0]
```

````python
inventory_lc = [
    {"Name": "Socks",  "ID": "1", "Price":  5.0, "Quantity": 10},
    {"Name": "Shoes",  "ID": "2", "Price": 20.0, "Quantity":  0},
    {"Name": "Gloves", "ID": "3", "Price":  8.0, "Quantity": 20},
    {"Name": "Hat",    "ID": "4", "Price": 12.0, "Quantity":  0},
]

# Extract all product names
names = [item["Name"] for item in inventory_lc]
print("All products:", names)
# Output: ['Socks', 'Shoes', 'Gloves', 'Hat']

# Names of products that are in stock (Quantity > 0)
in_stock = [item["Name"] for item in inventory_lc if item["Quantity"] > 0]
print("In stock:", in_stock)
# Output: ['Socks', 'Gloves']

# Prices of products costing more than $10
premium = [item["Price"] for item in inventory_lc if item["Price"] > 10]
print("Prices above $10:", premium)
# Output: [20.0, 12.0]

# All product IDs
ids = [item["ID"] for item in inventory_lc]
print("All IDs:", ids)
# Output: ['1', '2', '3', '4']
````

## StockMaster Inc. — Inventory Management System

Build the inventory system step by step. Run each cell in order and test it before moving on.

> Each step tells you which loop type to use. The concepts are the same ones > covered in this lesson — `while` loops, `for` loops, and loops over lists of dictionaries.

### Step 1: Initialise Inventory

Create an empty list. Each product will be added as a dictionary with four fields: `Name`, `ID`, `Price`, and `Quantity`.

````python
# Step 1: Initialise Inventory
inventory = []
````

### Step 2: Add Products (while loop)

Use a `while` loop to let the manager add products one at a time. Continue until the manager enters `"n"`.

````python
# Step 2: Add Products
while True:
    add = input("Add a new product? (y/n): ")
    if add.lower() == "n":
        break
    name     = input("  Product name: ")
    pid      = input("  Product ID: ")
    price    = float(input("  Price: "))
    quantity = int(input("  Quantity: "))
    inventory.append({"Name": name, "ID": pid, "Price": price, "Quantity": quantity})
    print(f"  '{name}' added.")
````

### Step 3: View Inventory (for loop)

Use a `for` loop to display every product currently in the inventory. Run this cell after adding products in Step 2.

````python
# Step 3: View Inventory
if not inventory:
    print("Inventory is empty.")
else:
    print("\n--- Current Inventory ---")
    for item in inventory:
        print(f"  {item['Name']} | ID: {item['ID']} | Price: ${item['Price']:.2f} | Qty: {item['Quantity']}")
    print()
````

### Step 4: Update Stock (for loop)

Use a `for` loop to update the quantity of multiple products. Prompt for how many products to update, then loop that many times.

````python
# Step 4: Update Stock
n_updates = int(input("How many products do you want to update? "))
for i in range(n_updates):
    pid     = input(f"  Update {i + 1}: Enter Product ID: ")
    new_qty = int(input("  New quantity: "))
    found = False
    for item in inventory:
        if item["ID"] == pid:
            item["Quantity"] = new_qty
            print(f"  Updated '{item['Name']}' to {new_qty} units.")
            found = True
            break
    if not found:
        print(f"  Product ID '{pid}' not found.")
````

### Step 5: Sales Transactions (loop over list of dicts)

Loop over a shopping cart (a list of dicts). For each cart item, find the matching product in the inventory and deduct the sold quantity.

> Update the sample cart below to match the Product IDs you created in Step 2.

````python
# Step 5: Sales Transactions
cart = [
    {"ID": "1", "Quantity": 2},
    {"ID": "2", "Quantity": 1},
]

for sold in cart:
    for item in inventory:
        if item["ID"] == sold["ID"]:
            if sold["Quantity"] > item["Quantity"]:
                print(f"  Insufficient stock for '{item['Name']}'.")
            else:
                item["Quantity"] -= sold["Quantity"]
                print(f"  Sold {sold['Quantity']}x '{item['Name']}'.")
            break
````

### Step 6: Restock Products (while loop)

Use a `while` loop to let the manager restock products by Product ID. Continue until the manager enters `"n"`.

````python
# Step 6: Restock Products
while True:
    restock = input("Restock a product? (y/n): ")
    if restock.lower() == "n":
        break
    pid     = input("  Product ID to restock: ")
    add_qty = int(input("  Quantity to add: "))
    found = False
    for item in inventory:
        if item["ID"] == pid:
            item["Quantity"] += add_qty
            print(f"  Restocked '{item['Name']}'. New quantity: {item['Quantity']}.")
            found = True
            break
    if not found:
        print(f"  Product ID '{pid}' not found.")
````

### Step 7: Final Inventory Display

Display the final state of the inventory and confirm the program has completed.

````python
# Step 7: Final Inventory Display
print("=== Final Inventory ===")
if not inventory:
    print("Inventory is empty.")
else:
    print("\n--- Final Inventory ---")
    for item in inventory:
        print(f"  {item['Name']} | ID: {item['ID']} | Price: ${item['Price']:.2f} | Qty: {item['Quantity']}")
    print()
print("Program completed successfully.")
````
