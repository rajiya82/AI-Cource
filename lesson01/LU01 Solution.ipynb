## Worked Example — Eligibility Calculator

*(Reference: 0104 - Worked Example)*

Evaluates a candidate's eligibility for a job based on age and gender criteria:
- General age range: 18–70
- Males: 18–65
- Females: 18–62

```python
# Input
age = int(input("Enter your age: "))
gender = input("Enter your gender (male/female): ").lower()

# Processing and Output
if 18 <= age <= 70:
    if gender == 'male':
        if 18 <= age <= 65:
            print("Eligible")
        else:
            print("Not Eligible")
    elif gender == 'female':
        if 18 <= age <= 62:
            print("Eligible")
        else:
            print("Not Eligible")
    else:
        print("Invalid gender input")
else:
    print("Not Eligible")
```

## Problem Statement — VO2Max Calculator

*(Reference: 0105 - Problem Statement)*

Calculates an athlete's VO2Max and determines suitability for a competitive running programme.

**Formulas:**
- `HR_max = 220 - age`
- `VO2Max = 15.3 × (HR_max / HR_rest)`

**Eligibility (age 18–39 only):**
- Males: VO2Max 60–85 mL/kg·min
- Females: VO2Max 50–75 mL/kg·min

> **Note:** The code below includes enhancements not described in the problem statement, such as handling invalid gender input and displaying the calculated VO2Max in all output messages. This is common in practice — when enhancements are made, document them clearly.

```python
import datetime

# Input
hr_rest = float(input("Enter your Resting Heart Rate (HR_rest): "))
year_of_birth = int(input("Enter your Year of Birth: "))
gender = input("Enter your Gender (male/female): ").lower()

# Processing
current_year = datetime.date.today().year
age = current_year - year_of_birth
hr_max = 220 - age
vo2_max = round(15.3 * (hr_max / hr_rest), 2)

# Output
if 18 <= age <= 39:
    if gender == 'male':
        if 60 <= vo2_max <= 85:
            print("VO2 Max is OK for Age & Gender. VO2 Max =", vo2_max)
        else:
            print("VO2 Max is Outside Range for Age & Gender. VO2 Max =", vo2_max)
    elif gender == 'female':
        if 50 <= vo2_max <= 75:
            print("VO2 Max is OK for Age & Gender. VO2 Max =", vo2_max)
        else:
            print("VO2 Max is Outside Range for Age & Gender. VO2 Max =", vo2_max)
    else:
        print("Invalid gender input.")
else:
    print("Age is outside range. VO2 Max =", vo2_max)
```
